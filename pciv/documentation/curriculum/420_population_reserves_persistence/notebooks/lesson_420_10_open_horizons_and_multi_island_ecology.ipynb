# Lesson 420.10: Multi-Island Migration Workbench

**Level:** Level 10 (Open Horizons & Multi-Island Ecology)  

This workbench demonstrates island migration and homologous crossover boundaries.

```python
class IslandEcology:
    def __init__(self, num_islands=3):
        self.islands = [[] for _ in range(num_islands)]
    
    def migrate(self, source_idx, target_idx, genome_id):
        assert genome_id in self.islands[source_idx], 'Genome must exist on source island'
        self.islands[source_idx].remove(genome_id)
        self.islands[target_idx].append(genome_id)

eco = IslandEcology(3)
eco.islands[0].append('G_alpha')
eco.migrate(0, 1, 'G_alpha')
assert 'G_alpha' in eco.islands[1]
print('Island Migration Mechanics Verified.')
```
