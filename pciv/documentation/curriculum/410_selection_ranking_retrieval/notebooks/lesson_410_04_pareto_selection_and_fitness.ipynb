This workbench verifies non-dominated Pareto front calculation using exact `fractions.Fraction`.

```python
from fractions import Fraction

def dominates(cand_a, cand_b):
    # cand = (accuracy, compliance)
    ge_all = all(a >= b for a, b in zip(cand_a, cand_b))
    gt_one = any(a > b for a, b in zip(cand_a, cand_b))
    return ge_all and gt_one

c1 = (Fraction(4, 5), Fraction(9, 10))
c2 = (Fraction(3, 5), Fraction(9, 10))
assert dominates(c1, c2), 'c1 must dominate c2'
print('Exact Rational Pareto Dominance Verified.')
```
