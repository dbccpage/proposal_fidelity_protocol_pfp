# Lesson 410.05: Inverse Lineage PageRank Workbench

**AuthorityEffect:** None.  
**Level:** Level 05 (Inverse Lineage PageRank)  

This workbench demonstrates backward Markov diffusion along inheritance DAG edges.

```python
from fractions import Fraction

def compute_ancestral_credit(children_scores, damping=Fraction(85, 100)):
    # Credit flows to parent based on children scores
    total_child_score = sum(children_scores, Fraction(0, 1))
    return damping * total_child_score

scores = [Fraction(1, 2), Fraction(1, 4)]
credit = compute_ancestral_credit(scores)
assert credit == Fraction(255, 400)
print(f'Ancestral PageRank Credit (Rational): {credit}')
```
