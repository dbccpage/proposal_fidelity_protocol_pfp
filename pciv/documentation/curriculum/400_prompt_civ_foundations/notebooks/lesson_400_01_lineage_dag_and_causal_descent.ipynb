**Level:** Level 01 (Lineage DAG & Causal Descent)  

This notebook verifies two-graph separation between the Inheritance DAG and Reference Graph.

```python
class LineageTopology:
    def __init__(self):
        self.inheritance_dag = {}
        self.reference_graph = {}
    
    def add_inheritance_edge(self, parent, child, edge_kind):
        assert edge_kind in ['DirectMutation', 'Repair', 'Retry'], 'Invalid inheritance edge'
        self.inheritance_dag.setdefault(child, []).append(parent)
        
    def add_reference_edge(self, source, target):
        # Reference citations stay strictly outside inheritance_dag
        self.reference_graph.setdefault(source, []).append(target)

topo = LineageTopology()
topo.add_inheritance_edge('P0', 'P1', 'DirectMutation')
topo.add_reference_edge('P1', 'P_ref')
assert 'P_ref' not in topo.inheritance_dag['P1'], 'F-LINEAGE-01: Reference edge must not pollute inheritance DAG'
print('Two-Graph Separation Verified.')
```
