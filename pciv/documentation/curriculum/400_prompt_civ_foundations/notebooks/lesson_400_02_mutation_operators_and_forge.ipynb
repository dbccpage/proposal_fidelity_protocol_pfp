This workbench verifies protected span rejection under `F-BIO-03`.

```python
def apply_mutation(prompt_bytes, span_start, span_end, new_text, protected_spans):
    for p_start, p_end in protected_spans:
        if max(span_start, p_start) < min(span_end, p_end):
            raise ValueError('PROTECTED_SPAN_MUTATION_REFUSED')
    return prompt_bytes[:span_start] + new_text + prompt_bytes[span_end:]

prompt = 'SYSTEM: AuthorityEffect: None. USER: Plastic text here.'
protected = [(0, 31)] # 'SYSTEM: AuthorityEffect: None.'

try:
    apply_mutation(prompt, 5, 15, 'MUTATED', protected)
    print('Error: Rejection failed!')
except ValueError as e:
    print(f'Forge Rejection Confirmed: {e}')
```
