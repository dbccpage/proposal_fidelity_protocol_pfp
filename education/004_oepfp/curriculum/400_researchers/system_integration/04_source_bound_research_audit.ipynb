```python
import importlib.util
import json
import sys
from pathlib import Path

ROOT = next(
    path
    for path in [Path.cwd(), *Path.cwd().parents]
    if (path / "education/scripts/build_system_integration.py").is_file()
)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))


def load_source(name, relative):
    spec = importlib.util.spec_from_file_location(name, ROOT / relative)
    module = importlib.util.module_from_spec(spec)
    sys.modules[name] = module
    spec.loader.exec_module(module)
    return module
```

## Worked computation

Run the cell, then compare the output with the exact model above.


```python
inventory_path = (
    ROOT
    / "education/documentation/004_oepfp/curriculum/research_inventory.json"
)
inventory = json.loads(inventory_path.read_text(encoding="utf-8"))
entries = inventory["files"]
paths = [entry["path"] for entry in entries]
assert len(paths) == len(set(paths)) and paths == sorted(paths)
allowed = {"lab-backed", "reading-only", "historical", "open"}
assert all(entry["classification"] in allowed and entry["learning_route"] for entry in entries)
assert all(len(entry["sha256"]) == 64 for entry in entries)
assert all(entry["lab_evidence"] for entry in entries if entry["classification"] == "lab-backed")
level_engine = [entry for entry in entries if entry["family"].startswith("level_engine/")]
assert {entry["owner"] for entry in level_engine} == {
    "symbol_game",
    "operations",
    "governance",
    "lean4",
}
history_prefix = "symbol_game/infrastructure/tower_evidence/level_engine/v1/"
assert all(
    entry["classification"] == "historical"
    for entry in level_engine
    if entry["path"].startswith(history_prefix)
)
assert not any(entry["path"].startswith("research/level_engine/") for entry in entries)
print("Inventory entries:", len(entries), "classification counts:", inventory["counts"])
```

## Exercise and adversarial check

Check one retained table source hash, then alter its bytes in memory and detect the mismatch. Do not modify the source file.

Predict the result before running the reference check below. Each assertion checks a specific value or failure mode.


```python
import hashlib

anchor = next(
    entry
    for entry in entries
    if entry["path"]
    == "symbol_game/infrastructure/tower_research/reach/data/stages.tsv"
)
raw = (ROOT / anchor["path"]).read_bytes()
assert hashlib.sha256(raw).hexdigest() == anchor["sha256"]
assert hashlib.sha256(raw + b"changed").hexdigest() != anchor["sha256"]
assert anchor["classification"] == "lab-backed"
assert any(entry["classification"] == "reading-only" for entry in entries)
print("A changed source digest invalidates the old execution claim.")
```
