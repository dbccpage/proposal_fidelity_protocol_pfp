## Worked computation

Run the cell, then compare the output with the exact model above.


```python
from collections import Counter

from omega_engine_pfp.domain.tower import registry
from omega_engine_pfp.domain.tower.stage_status import (
    BINDINGS,
    GRADES,
    census,
    stage_status,
)

stages = registry.load_stages()
registry.check_registry()
assert [stage.level for stage in stages] == list(range(101))
counts = Counter(stage.status for stage in stages)
assert set(counts) <= set(GRADES)
assert counts == Counter({grade: len(levels) for grade, levels in census().items() if levels})
assert all(stage.obligation and stage.counterexample for stage in stages)
reserved = [opcode for stage in stages for opcode in stage.reserved_opcodes]
assert len(reserved) == len(set(reserved))
print("Computed status counts:", dict(sorted(counts.items())))
```

## Exercise and adversarial check

Inspect stages 15, 16, 21, 22, 35 and 100. Check that each status is computed from evidence, not read off the stage number.

Predict the result before running the reference check below. Each assertion checks a specific value or failure mode.


```python
assert all(stage.status == stage_status(stage.level).grade for stage in stages)
computed = [stage.level for stage in stages if stage.status == "COMPUTED"]
assert all(any(binding.level == level for binding in BINDINGS) for level in computed)
print("Stages whose registered checker holds:", computed)
assert not (set(reserved) & registry.SOURCE_OPCODE_SET)
accepted_lift_records = []  # Declared lesson fixture, not a live external-repository query.
assert len(accepted_lift_records) == 0
assert len(stages) != len(accepted_lift_records)
print("An indexed level and a licensed lift record are different evidence objects.")
```
