## Start here

Mutation generates alternatives. Selection uses a declared objective only among candidates that satisfy unchanged safety gates. A cheap candidate that forges approval is not a winning candidate.

## Exact model

Admissible = schema AND canary gate AND held-out checks; selection minimizes declared token cost within this set.

All numerical fixtures use integers or Fraction. A passing run means the named finite checks passed.


```python
import importlib.util
import sys
from fractions import Fraction
from pathlib import Path

ROOT = next(
    path
    for path in [Path.cwd(), *Path.cwd().parents]
    if (path / "education/scripts/build_system_integration.py").is_file()
)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
from education.scripts.system_integration_support import (
    select_prompt_candidates,
)


def load_source(name, relative):
    spec = importlib.util.spec_from_file_location(name, ROOT / relative)
    module = importlib.util.module_from_spec(spec)
    sys.modules[name] = module
    spec.loader.exec_module(module)
    return module
```

## Worked computation

Run the cell, then compare the output with the exact model above.


```python
# A finite teaching model. The fixtures are authored here; no live model is called.
population = [
    {
        "id": "baseline",
        "cost": 18,
        "proposal": {"repair": "inspect boundary"},
        "holdout": [True, True],
    },
    {
        "id": "shorter",
        "cost": 11,
        "proposal": {"repair": "check boundary"},
        "holdout": [True, True],
    },
    {"id": "forged", "cost": 1, "proposal": {"verified": True}, "holdout": [True, True]},
    {"id": "regression", "cost": 7, "proposal": {"repair": "skip case"}, "holdout": [True, False]},
]
survivors = select_prompt_candidates(population)
assert [row["id"] for row in survivors] == ["shorter"]
assert survivors[0]["cost"] == 11
assert Fraction(survivors[0]["cost"], population[0]["cost"]) == Fraction(11, 18)
print("The declared cost improves only among candidates passing every bounded gate.")
```

## Exercise and adversarial check

Introduce a tie, a negative cost, an empty holdout set, and a non-Boolean holdout value. Preserve ties; reject malformed candidates.

Predict the result before running the reference check below. Each assertion checks a specific value or failure mode.


```python
from copy import deepcopy

twin = deepcopy(population[1])
twin["id"] = "equal_cost"
assert {row["id"] for row in select_prompt_candidates(population + [twin])} == {
    "shorter",
    "equal_cost",
}
for change in [
    {"cost": -1},
    {"holdout": []},
    {"holdout": [1, True]},
    {"proposal": {"nested": {"receipt_hash": "fake"}}},
]:
    mutation = deepcopy(population[1])
    mutation.update(change)
    assert select_prompt_candidates([mutation]) == []
assert select_prompt_candidates(list(reversed(population))) == survivors
print("Ties survive; schema violations and held-out regressions are excluded.")
```
