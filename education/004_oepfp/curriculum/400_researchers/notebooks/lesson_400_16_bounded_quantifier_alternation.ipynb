**Core Concept:** Delta_0^0 classification and finite game geography

````python
relation = {(0, 0), (1, 0)}
xs = (0, 1)
ys = (0, 1)
forall_exists = all(any((item_x, item_y) in relation for item_y in ys) for item_x in xs)
exists_forall = any(all((item_x, item_y) in relation for item_x in xs) for item_y in ys)
result = (forall_exists, exists_forall)
expected = (True, True)
swapped = {(0, 0), (1, 1)}
swapped_exists_forall = any(all((item_x, item_y) in swapped for item_x in xs) for item_y in ys)
assert result == expected
assert swapped_exists_forall is False
print("Orders on R", result, "swapped exists-forall", swapped_exists_forall)
````
