## 1. The Big Idea

**Core Concept:** Enforcing zero unpriced rewrite density across multi-agent swarms

````python
rewrites = ({"priced": True}, {"priced": False})
unpriced = sum(1 for item in rewrites if not item["priced"])
result = "REFUSED" if unpriced else "ADMITTED"
expected = "REFUSED"
assert unpriced == 1
assert result == expected
print("Swarm gate", result)
````
