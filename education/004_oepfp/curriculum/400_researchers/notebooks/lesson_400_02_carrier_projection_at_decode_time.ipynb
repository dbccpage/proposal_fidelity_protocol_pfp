## 1. The Big Idea

**Core Concept:** Forcing 0.04% -> 100% on-carrier mass via negative infinity masking

````python
from fractions import Fraction

scores = {"<a>": Fraction(1), "fluff": Fraction(2), "<b>": Fraction(3)}
carrier = {token: value for token, value in scores.items() if token.startswith("<")}
total = sum(carrier.values())
result = total / total
expected = Fraction(1)
assert sum(scores.values()) == 6
assert total == 4
assert result == expected
print("On-carrier mass after the mask", result)
````
