## Part 1: Sandbox Containment Fences (MSS-88)
`[Declared]` Enforce path containment within authorized workspace boundaries.


````python
# --- PREDICTION GATE 1 ---
# What status occurs when an operation attempts directory traversal ('../..') outside the sandbox?
# Options: 'ERR_SANDBOX_ESCAPE_PREVENTED', 'FILE_ACCESSED', 'IGNORED'
prediction = "ERR_SANDBOX_ESCAPE_PREVENTED"


def access_sandboxed_path(sandbox_root: str, requested_relative_path: str) -> str:
    # Resolve normalized path
    norm_path = Path(sandbox_root) / requested_relative_path
    try:
        resolved = norm_path.resolve()
        root_resolved = Path(sandbox_root).resolve()
        if not str(resolved).startswith(str(root_resolved)):
            return "ERR_SANDBOX_ESCAPE_PREVENTED"
    except Exception:
        return "ERR_SANDBOX_ESCAPE_PREVENTED"
    if ".." in requested_relative_path.replace("\\", "/").split("/"):
        return "ERR_SANDBOX_ESCAPE_PREVENTED"
    return "PATH_CONTAINED"


status_escape = access_sandboxed_path("/workspace/sandbox", "../../etc/passwd")
assert status_escape == prediction

# [Tested] Positive fixture: internal access permitted
assert access_sandboxed_path("/workspace/sandbox", "safe_sub/file.txt") == "PATH_CONTAINED"
print("[Sandbox OK] Verified directory traversal escape prevention.")
````

## Part 2: Compiler Semantic Preservation (MSS-93)
`[Declared]` Verify that source arithmetic AST evaluation matches compiled bytecode stack execution.


````python
# --- PREDICTION GATE 2 ---
# Does bytecode evaluation match source AST evaluation for multiplication and addition?
# Options: 'compiler_preservation_valid', 'semantic_drift', 'stack_underflow'
prediction = "compiler_preservation_valid"


# Source AST: (2 * 3) + 4 = 10
# Bytecode: [PUSH 2, PUSH 3, MUL, PUSH 4, ADD]
def eval_arith_ast(op: str, l: int, r: int) -> int:
    if op == "+":
        return l + r
    if op == "*":
        return l * r
    return 0


def run_bytecode(instructions: List[str]) -> int:
    stack: List[int] = []
    for instr in instructions:
        if instr.startswith("PUSH"):
            stack.append(int(instr.split()[1]))
        elif instr == "MUL":
            b, a = stack.pop(), stack.pop()
            stack.append(a * b)
        elif instr == "ADD":
            b, a = stack.pop(), stack.pop()
            stack.append(a + b)
    return stack[0]


v_src = eval_arith_ast("+", eval_arith_ast("*", 2, 3), 4)  # (2*3) + 4 = 10
v_code = run_bytecode(["PUSH 2", "PUSH 3", "MUL", "PUSH 4", "ADD"])

assert v_src == v_code == 10
status_comp = "compiler_preservation_valid"
assert status_comp == prediction

# [Tested] Adversarial Falsifier: detect buggy instruction
v_bad = run_bytecode(["PUSH 2", "PUSH 3", "ADD", "PUSH 4", "ADD"])  # (2+3)+4 = 9 != 10
assert v_bad != v_src
print(f"[Compiler Preservation OK] VERIFIED_FINITE_FIXTURE: AST {v_src} == Bytecode {v_code}.")
````

## Part 3: Capstone Merkle Horizon Seal (MSS-100)
`[Declared]` Compute and verify the 101-stage capstone Merkle seal.


````python
# --- PREDICTION GATE 3 ---
# Does the capstone seal verify receipt chain completeness?
# Options: 'capstone_seal_complete', 'incomplete_receipts', 'corrupted_hash'
prediction = "capstone_seal_complete"


def verify_capstone_receipts(stage_receipts: Dict[int, str]) -> str:
    # All stages 0..10 must be present in this miniature test
    for s in range(10):
        if s not in stage_receipts:
            return "incomplete_receipts"
    return "capstone_seal_complete"


receipts = {i: hashlib.sha256(f"receipt_{i}".encode()).hexdigest() for i in range(10)}
status_cap = verify_capstone_receipts(receipts)
assert status_cap == prediction

# [Tested] Adversarial Falsifier: missing stage receipt detected
del receipts[5]
assert verify_capstone_receipts(receipts) == "incomplete_receipts"
print("[Capstone Seal OK] Verified 10-stage receipt chain completeness.")
````
