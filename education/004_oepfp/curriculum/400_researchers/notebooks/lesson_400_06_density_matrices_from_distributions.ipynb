## Part 1: Finite Markov Kernels in $\mathbb{Q}$
`[Declared]` Verify row-stochasticity and exact matrix multiplication for finite state transitions.


````python
# --- PREDICTION GATE 1 ---
# Does every row in a valid Markov transition kernel sum to exactly Fraction(1, 1)?
# Options: 'stochastic_row_sum_one', 'approximate_one', 'unconstrained'
prediction = "stochastic_row_sum_one"


def verify_stochastic_matrix(matrix: List[List[Fraction]]) -> str:
    for row in matrix:
        for p in row:
            if p < 0:
                return "ERR_NEGATIVE_PROBABILITY"
        if sum(row) != Fraction(1):
            return "ERR_ROW_SUM_MISMATCH"
    return "stochastic_row_sum_one"


# Valid 2x2 stochastic matrix in Q
P = [[Fraction(2, 3), Fraction(1, 3)], [Fraction(1, 2), Fraction(1, 2)]]
status = verify_stochastic_matrix(P)
assert status == prediction

# [Tested] Adversarial Falsifier: reject defective non-normalized matrix
P_bad = [[Fraction(2, 3), Fraction(2, 3)], [Fraction(1, 2), Fraction(1, 2)]]
assert verify_stochastic_matrix(P_bad) == "ERR_ROW_SUM_MISMATCH"
print("[Markov Kernel OK] Verified row-stochastic normalization in exact rational arithmetic.")
````

## Part 2: Exact Rational Bayesian Inference
`[Declared]` Compute exact posterior probabilities without floating-point approximations.


````python
# --- PREDICTION GATE 2 ---
# Given prior P(H)=1/3 and likelihoods P(E|H)=3/4, P(E|~H)=1/4, what is posterior P(H|E)?
# Options: 'bayes_posterior_normalized', 'p_3_5', 'p_1_2'
prediction = "bayes_posterior_normalized"


def exact_bayes_update(
    prior_h: Fraction, p_e_given_h: Fraction, p_e_given_not_h: Fraction
) -> Tuple[str, Fraction]:
    prior_not_h = Fraction(1) - prior_h
    marginal_e = p_e_given_h * prior_h + p_e_given_not_h * prior_not_h
    if marginal_e == 0:
        return ("ERR_ZERO_EVIDENCE", Fraction(0))
    posterior_h = (p_e_given_h * prior_h) / marginal_e
    return ("bayes_posterior_normalized", posterior_h)


status, post = exact_bayes_update(Fraction(1, 3), Fraction(3, 4), Fraction(1, 4))
assert status == prediction
assert post == Fraction(3, 5), f"Expected 3/5, got {post}"

# [Tested] Adversarial Falsifier: zero marginal evidence is caught cleanly
status_err, _ = exact_bayes_update(Fraction(1), Fraction(0), Fraction(0))
assert status_err == "ERR_ZERO_EVIDENCE"
print(f"[Bayes OK] Verified exact rational posterior = {post}.")
````

## Part 3: Linear Resource Token Preservation
`[Declared]` Enforce that linear capability tokens cannot be duplicated or discarded without audit.


````python
# --- PREDICTION GATE 3 ---
# Can a linear capability token be consumed more than once?
# Options: 'resource_conserved', 'double_spend_allowed', 'unbounded'
prediction = "resource_conserved"


class LinearResourceVault:
    def __init__(self, initial_tokens: int):
        self.tokens = initial_tokens

    def consume_token(self) -> str:
        if self.tokens <= 0:
            return "ERR_RESOURCE_TOKEN_DEFECT"
        self.tokens -= 1
        return "resource_conserved"


vault = LinearResourceVault(1)
res1 = vault.consume_token()
assert res1 == prediction

# [Tested] Adversarial Falsifier: second consumption attempt must trigger typed defect
res2 = vault.consume_token()
assert res2 == "ERR_RESOURCE_TOKEN_DEFECT"
print("[Linear Resources OK] Verified single-use token conservation.")
````
