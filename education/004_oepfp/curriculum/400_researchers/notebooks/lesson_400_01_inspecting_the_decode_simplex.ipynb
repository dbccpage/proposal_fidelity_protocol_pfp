**Core Concept:** Analyzing unnormalized logits, softmax simplices, and logprobs

````python
from fractions import Fraction

scores = (Fraction(1), Fraction(2), Fraction(3))
total = sum(scores)
weights = tuple(score / total for score in scores)
result = weights[1]
expected = Fraction(1, 3)
assert weights == (Fraction(1, 6), Fraction(1, 3), Fraction(1, 2))
assert result == expected
print("Simplex weight", result)
````
