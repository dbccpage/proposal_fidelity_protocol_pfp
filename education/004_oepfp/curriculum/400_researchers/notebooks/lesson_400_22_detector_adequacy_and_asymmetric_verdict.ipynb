````python
red = {"verdict": "RED", "witness": "cycle-3"}
green = {"verdict": "GREEN", "witness": None}


def keep(report):
    if report["verdict"] == "RED" and report["witness"] is not None:
        return "RED"
    if report["verdict"] == "GREEN" and report["witness"] is None:
        return "NOT_ADMITTED"
    return "UNSCORED"


result = (keep(red), keep(green))
expected = ("RED", "NOT_ADMITTED")
assert result == expected
print("Detector filter", result)
````
