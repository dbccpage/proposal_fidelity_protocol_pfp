## 1. The Big Idea

**Core Concept:** Exact rational planar angle sums and turn fractions

### The Four-Tier Perspective

````python
from fractions import Fraction

turns = tuple(Fraction(1, 6) for _ in range(5))
result = Fraction(1) - sum(turns)
expected = Fraction(1, 6)
assert sum(tuple(Fraction(1, 6) for _ in range(6))) == 1
assert result == expected
print("Turn deficit", result)
````
