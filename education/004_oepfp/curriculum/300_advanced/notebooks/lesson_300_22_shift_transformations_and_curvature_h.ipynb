## 1. The Big Idea

**Core Concept:** Non-abelian 2-curvature H = d_2 B + CS(A)

### The Four-Tier Perspective
- **Tangible (Age 5–10):** Physical intuition of shift transformations and curvature h.

- **Discrete Math (Undergrad):** Exact rational invariants and coboundary calculus.
- **Foundational (Age 50–500):** Monotone manifold retraction and topological obstruction theory.


````python
from fractions import Fraction

cochain = {(0, 1): Fraction(1), (1, 2): Fraction(2), (2, 0): Fraction(3)}
shift = {0: Fraction(0), 1: Fraction(1), 2: Fraction(0)}
shifted = {
    (source, target): value + shift[target] - shift[source]
    for (source, target), value in cochain.items()
}
result = sum(shifted.values())
expected = sum(cochain.values())
assert expected == 6
assert result == expected
print("Circulation after the shift", result)
````
