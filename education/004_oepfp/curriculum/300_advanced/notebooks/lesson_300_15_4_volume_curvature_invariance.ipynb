## Worked computation

````python
# Not runnable: domain.regge has no successor in omega_engine_pfp.

T = [
    [Fraction(0), Fraction(1), Fraction(0)],
    [Fraction(0), Fraction(0), Fraction(1)],
    [Fraction(1), Fraction(0), Fraction(0)],
]
identity = [[Fraction(i == j) for j in range(3)] for i in range(3)]
power = identity
traces = []
for n in range(1, 10):
    power = matmul(power, T)
    trace = sum(power[i][i] for i in range(3))
    assert trace == (3 if n % 3 == 0 else 0)
    traces.append(trace)
u = Fraction(1, 2)
matrix = [[identity[i][j] - u * T[i][j] for j in range(3)] for i in range(3)]
assert _det(matrix) == determinant_leibniz(matrix) == 1 - u**3
print("Return counts:", traces, "det(I-uT):", _det(matrix))
````
