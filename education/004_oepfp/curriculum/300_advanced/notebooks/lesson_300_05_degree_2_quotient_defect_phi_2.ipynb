## 1. The Big Idea

**Core Concept:** Phi_2(B) = min_A ||B - d_1 A||_{1, W} on 2-cochains

### The Four-Tier Perspective

- **Algorithmic (Age 11–17):** Deterministic state transitions and rule sets.
- **Discrete Math (Undergrad):** Exact rational invariants and coboundary calculus.
- **Foundational (Age 50–500):** Monotone manifold retraction and topological obstruction theory.


````python
from fractions import Fraction


def distance(multiple):
    target = (Fraction(1), Fraction(0), Fraction(0), Fraction(0))
    coboundary = tuple(multiple for _ in range(4))
    return sum(abs(target[index] - coboundary[index]) for index in range(4))


scores = {multiple: distance(Fraction(multiple)) for multiple in range(-2, 3)}
result = min(scores.values())
expected = Fraction(1)
assert scores[0] == 1
assert scores[1] == 3
assert result == expected
print("Phi2 on this declared coboundary line", result)
````
