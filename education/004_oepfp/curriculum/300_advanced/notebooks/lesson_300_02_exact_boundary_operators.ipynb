## 1. The Big Idea

**Core Concept:** Computing boundary matrices d_k and verifying d_{k-1} d_k = 0

### The Four-Tier Perspective

````python
edges = ((0, 1), (1, 2), (2, 0))
vertex = [0, 0, 0]
for source, target in edges:
    vertex[source] -= 1
    vertex[target] += 1
result = tuple(vertex)
expected = (0, 0, 0)
assert result == expected
print("Boundary of boundary", result)
````
