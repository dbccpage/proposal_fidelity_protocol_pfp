## 1. The Big Idea

**Core Concept:** Computing exact rational Cayley-Menger determinants for simplices

### The Four-Tier Perspective
- **Tangible (Age 5–10):** Physical intuition of metric simplices and cayley-menger.
- **Algorithmic (Age 11–17):** Deterministic state transitions and rule sets.
- **Discrete Math (Undergrad):** Exact rational invariants and coboundary calculus.
- **Foundational (Age 50–500):** Monotone manifold retraction and topological obstruction theory.


````python
from fractions import Fraction


def det_over_q(matrix):
    size = len(matrix)
    acc = [[Fraction(entry) for entry in row] for row in matrix]
    sign = Fraction(1)
    for col in range(size):
        pivot = next((r for r in range(col, size) if acc[r][col] != 0), None)
        if pivot is None:
            return Fraction(0)
        if pivot != col:
            acc[col], acc[pivot] = acc[pivot], acc[col]
            sign = -sign
        scale = acc[col][col]
        for other in range(col + 1, size):
            factor = acc[other][col] / scale
            for cursor in range(col, size):
                acc[other][cursor] -= factor * acc[col][cursor]
    diagonal = sign
    for col in range(size):
        diagonal *= acc[col][col]
    return diagonal


squared = (Fraction(9), Fraction(16), Fraction(25))
matrix = [
    [0, 1, 1, 1],
    [1, 0, squared[0], squared[1]],
    [1, squared[0], 0, squared[2]],
    [1, squared[1], squared[2], 0],
]
cayley = -det_over_q(matrix)
heron = Fraction(16) * Fraction(36)
result = cayley
expected = Fraction(576)
assert result == heron == expected
print("16 area^2", result)
````

## Part 1: Exact Rational Cayley-Menger Volume Positivity
`[Tested]` Objective: Certify 2-simplex embeddability in Q.


````python
# --- PREDICTION GATE 1 ---
# What is the CM determinant sign for a degenerate flat triangle with lengths (1, 2, 3)?
# Options: '0', 'Positive', 'Negative'
prediction = "0"


def cm_det(s01: Fraction, s12: Fraction, s02: Fraction) -> Fraction:
    # 2*s01*s12 + 2*s12*s02 + 2*s01*s02 - s01^2 - s12^2 - s02^2
    return 2 * s01 * s12 + 2 * s12 * s02 + 2 * s01 * s02 - s01**2 - s12**2 - s02**2


# Degenerate collinear case: lengths 1, 2, 3 (squared: 1, 4, 9)
det_collinear = cm_det(Fraction(1), Fraction(4), Fraction(9))
assert str(det_collinear) == prediction
print(f"[Tested OK] Collinear triangle CM determinant vanishes exactly: {det_collinear}")
````

## Part 2: Hinge Deficit Angle Calculation
`[Tested]` Objective: Verify discrete curvature vanishing on flat interior vertices.


````python
# --- PREDICTION GATE 2 ---
# What is the deficit angle around a vertex surrounded by 6 equilateral triangles?
# Options: '0', '2pi', 'pi'
prediction = "0"


# [Tested] Adversarial Falsifier & Counterexample: non-flat deficit angle
# In turn fractions: 6 triangles * 1/6 turn = 1 full turn (2pi). Deficit = 1 - 1 = 0
def deficit_angle_turns(triangles_count: int) -> Fraction:
    # Each equilateral triangle has angle = 1/6 turn
    angle_sum = Fraction(triangles_count, 6)
    return Fraction(1) - angle_sum


flat_deficit = deficit_angle_turns(6)
assert str(flat_deficit) == prediction

# Falsifier: 5 triangles produces positive deficit (spherical cone)
curved_deficit = deficit_angle_turns(5)
assert curved_deficit == Fraction(1, 6)
print(f"[Tested OK] Flat deficit = {flat_deficit}, Curved deficit = {curved_deficit} turns")
````

## Part 3: Cumulative Simplicial Curvature Integration
`[Declared]` Objective: Verify total deficit angle sum across a closed topological 2-sphere triangulation.


````python
# --- PREDICTION GATE 3 ---
# In a tetrahedral triangulation of S^2, what is the sum of deficit angles across all 4 vertices (turns)?
# Options: '2', '0', '1'
prediction = "2"


# Concrete triangulation of closed 2-sphere S^2 via regular tetrahedron:
# Vertices V = 4, Edges E = 6, Faces F = 4
# Each face is an equilateral triangle with face angle = 1/6 turn (pi/3 radians)
# Exactly 3 faces meet at each vertex.
def compute_tetrahedral_gauss_bonnet() -> Fraction:
    V = 4
    angles_per_vertex = 3 * Fraction(1, 6)  # 1/2 turn
    deficit_per_vertex = Fraction(1) - angles_per_vertex  # 1/2 turn
    total_deficit = V * deficit_per_vertex  # 4 * 1/2 = 2 turns
    return total_deficit


total_def = compute_tetrahedral_gauss_bonnet()
assert str(total_def) == prediction
# Euler characteristic check: 2pi * chi(S^2) = 2pi * (V - E + F) = 2pi * (4 - 6 + 4) = 4pi radians = 2 turns
assert total_def == Fraction(2)
print(
    f"[Tested OK] Discrete Gauss-Bonnet computed from tetrahedral triangulation: total deficit = {total_def} turns."
)
````
