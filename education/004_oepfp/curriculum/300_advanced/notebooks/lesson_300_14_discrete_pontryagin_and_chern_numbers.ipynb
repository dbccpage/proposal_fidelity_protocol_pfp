**Core Concept:** Evaluating <F cup F, [M^4]> pairings over discrete 4-manifolds

````python
from fractions import Fraction

left = (Fraction(1), Fraction(0), Fraction(0), Fraction(0))
right = (Fraction(1), Fraction(1), Fraction(0), Fraction(0))
result = sum(left[index] * right[index] for index in range(4))
expected = Fraction(1)
assert result == expected
print("Declared pairing", result)
````

## Exact model

For Q=[[0,1],[1,0]], x^T Q x=2ab. At x=(1,1) this is 2.

All numerical fixtures use integers or Fraction. A passing run means the named finite checks passed.


````python
import importlib.util
import json
import sys
from fractions import Fraction
from pathlib import Path

ROOT = next(
    path
    for path in [Path.cwd(), *Path.cwd().parents]
    if (path / "omega_engine_education/scripts/build_system_integration.py").is_file()
)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))


def load_source(name, relative):
    spec = importlib.util.spec_from_file_location(name, ROOT / relative)
    module = importlib.util.module_from_spec(spec)
    sys.modules[name] = module
    spec.loader.exec_module(module)
    return module
````

## Worked computation

Run the cell, then compare the output with the exact model above.


````python
degree4 = load_source(
    "degree4_lesson",
    "omega_engine_education/pilots/demo_003/demo_9/dossier_2_pontryagin_chern_4manifold/checker.py",
)
source_witness = json.loads(
    (
        ROOT
        / "omega_engine_education/pilots/demo_003/demo_9/dossier_2_pontryagin_chern_4manifold/witness_valid.json"
    ).read_text(encoding="utf-8")
)
assert source_witness["instanton_certificate"]["reported_value"] == "2"
Q = [[Fraction(0), Fraction(1)], [Fraction(1), Fraction(0)]]
def pair(a, b):
    return sum([a, b][i] * Q[i][j] * [a, b][j] for i in range(2) for j in range(2))
assert pair(Fraction(1), Fraction(1)) == 2
assert pair(Fraction(2), Fraction(3)) == 12
assert degree4.evaluate_pontryagin_pairing([Fraction(2)], [Fraction(1)]) == 2
d3, d2, d1, d0 = (
    [[Fraction(0), Fraction(0)]],
    [[Fraction(0), Fraction(0)]] * 2,
    [[Fraction(0)]] * 2,
    [[Fraction(0)]],
)
assert all(degree4.verify_chain_law_nilpotence(d3, d2, d1, d0))
certificate = degree4.check_exact_phi4_certificate(
    d3, d2, d1, d0, [Fraction(2)], [Fraction(0)] * 2, [Fraction(1)], Fraction(2)
)
assert certificate["status"] == "PASS" and certificate["duality_gap"] == "0"
print("Declared pairing and finite degree-four certificate both evaluate to two.")
````

````python
wrong = degree4.check_exact_phi4_certificate(
    d3, d2, d1, d0, [Fraction(2)], [Fraction(0)] * 2, [Fraction(1)], Fraction(3)
)
assert wrong["status"] == "REFUSED_REPORTED_VALUE_MISMATCH"
unbounded = degree4.check_exact_phi4_certificate(
    d3, d2, d1, d0, [Fraction(2)], [Fraction(0)] * 2, [Fraction(2)], Fraction(2)
)
assert unbounded["status"] == "REFUSED_DUAL_NORM_EXCEEDED"
assert not all(
    degree4.verify_chain_law_nilpotence(
        [[Fraction(1), Fraction(0)]],
        [[Fraction(1), Fraction(0)], [Fraction(0), Fraction(0)]],
        d1,
        d0,
    )
)
assert pair(Fraction(1), Fraction(-1)) == -2
print("Claim value, dual bound, orientation, and chain law remain separate checks.")
````
