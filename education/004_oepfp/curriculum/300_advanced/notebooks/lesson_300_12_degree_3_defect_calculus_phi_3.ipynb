**Core Concept:** Quantifying degree-3 curvature defects on hyper-prisms

````python
from fractions import Fraction


def distance(multiple):
    target = (Fraction(1), Fraction(0))
    image = (multiple, multiple)
    return abs(target[0] - image[0]) + abs(target[1] - image[1])


result = min(distance(Fraction(multiple)) for multiple in range(-2, 3))
expected = Fraction(1)
assert result == expected
print("Degree-3 residual on this two-face line", result)
````
