## Worked computation

Run the cell, then compare the output with the exact model above.


```python
from itertools import combinations
from math import factorial

# Not runnable: domain.regge has no successor in omega_engine_pfp.

vertices = list(range(5))
squared = {edge: Fraction(1) for edge in combinations(vertices, 2)}
determinant = regge_exact.cayley_menger_determinant(vertices, squared)
assert determinant == -5
volume_squared = -determinant / (2**4 * factorial(4) ** 2)
assert volume_squared == Fraction(5, 9216)
# Independent Leibniz determinant, rather than the source Gaussian elimination.
matrix = [[Fraction(0) if i == j else Fraction(1) for j in range(6)] for i in range(6)]
assert determinant_leibniz(matrix) == determinant
print("Computed determinant:", determinant, "volume squared:", volume_squared)
```
