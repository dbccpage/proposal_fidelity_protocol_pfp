## Worked computation

Run the cell, then compare the output with the exact model above.


```python
# Not runnable: domain.regge has no successor in pfp.

T = [
    [Fraction(0), Fraction(1), Fraction(0)],
    [Fraction(0), Fraction(0), Fraction(1)],
    [Fraction(1), Fraction(0), Fraction(0)],
]
identity = [[Fraction(i == j) for j in range(3)] for i in range(3)]
power = identity
traces = []
for n in range(1, 10):
    power = matmul(power, T)
    trace = sum(power[i][i] for i in range(3))
    assert trace == (3 if n % 3 == 0 else 0)
    traces.append(trace)
u = Fraction(1, 2)
matrix = [[identity[i][j] - u * T[i][j] for j in range(3)] for i in range(3)]
assert _det(matrix) == determinant_leibniz(matrix) == 1 - u**3
print("Return counts:", traces, "det(I-uT):", _det(matrix))
```

## Exercise and adversarial check

Remove the closing edge. Compare the determinant and return counts with the cycle.

Predict the result before running the reference check below. Each assertion checks a specific value or failure mode.


```python
broken = [row[:] for row in T]
broken[2][0] = Fraction(0)
assert matmul(matmul(broken, broken), broken) == [[Fraction(0)] * 3 for _ in range(3)]
broken_matrix = [[identity[i][j] - u * broken[i][j] for j in range(3)] for i in range(3)]
assert _det(broken_matrix) == 1
print("Removing the closing edge removes all positive-length returns.")
```
