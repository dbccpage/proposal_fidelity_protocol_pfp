## Part 1: S(n) Ground State & Void Presentation
`[Declared]` Objective: Verify Level 0 state space initialization.
Explicitly distinguish semantic payload capacity (0) from wire presentation token (`<>`).


````python
# --- PREDICTION GATE 1 ---
# What is the semantic payload capacity of the Ground State S(0)?
# Options: '0', '1', 'None'
prediction = "0"


@dataclass(frozen=True)
class SnState:
    level: int
    wire_token: str = "<>"

    @property
    def payload_capacity(self) -> int:
        # Semantic carrier size is 0: the presentation void encloses no payload
        return 0


s0 = SnState(level=0)
assert str(s0.payload_capacity) == prediction
assert s0.level == 0
assert s0.wire_token == "<>"
print(
    f"[Executed OK] S(0) Ground State confirmed: payload capacity = {s0.payload_capacity}, wire token = {s0.wire_token}"
)
````

## Part 2: Supply Derivation & Admissibility Filtration
`[Tested]` Objective: Filter candidate state extensions under admissibility constraints.


````python
# --- PREDICTION GATE 2 ---
# Will an off-carrier element ('<bad>') pass admissibility check at level 1?
# Options: 'False', 'True'
prediction = "False"


def is_admissible(elem: str, level: int) -> bool:
    if level == 0:
        return elem == "<>"
    if level == 1:
        return elem in ("<>", "<0>", "<1>")
    return False


test_pass = is_admissible("<bad>", level=1)
assert str(test_pass) == prediction
assert is_admissible("<0>", level=1) is True
print("[Tested OK] Admissibility filtration rejected off-carrier element.")
````

## Part 3: Lattice Core Collapse & Confluence
`[Tested]` Objective: Verify order-independent collapse across two distinct rewrite sequences.


````python
# --- PREDICTION GATE 3 ---
# Do distinct rewrite pathways from <<<0>>> converge to identical normal form <0>?
# Options: 'True', 'False'
prediction = "True"


def reduce_inner_first(term: str) -> str:
    # Rewrite pathway 1: inner wrap first -> <<0>> -> <0>
    return term.replace("<<<0>>>", "<<0>>").replace("<<0>>", "<0>")


def reduce_outer_first(term: str) -> str:
    # Rewrite pathway 2: outer wrap first
    t = term[1:-1] if term.startswith("<<") and term.endswith(">>") else term
    t = t[1:-1] if t.startswith("<<") and t.endswith(">>") else t
    return t


initial_term = "<<<0>>>"
nf1 = reduce_inner_first(initial_term)
nf2 = reduce_outer_first(initial_term)
assert nf1 == "<0>"
assert nf2 == "<0>"
assert (nf1 == nf2) == (prediction == "True")
print(f"[Tested OK] Core collapse confluence verified: different rewrite sequences yield {nf1}")
````

## Part 4: Metric Separation Axiom & Pseudometric Falsifier
`[Pinned]` Objective: Confirm that distance zero requires explicit metric separation before asserting point equality.


````python
# --- PREDICTION GATE 4 ---
# In a pseudometric space, does d(x, y) == 0 imply x == y without the separation axiom?
# Options: 'False', 'True'
prediction = "False"


# Demonstration: pseudometric on state vectors
class PseudometricSpace:
    def __init__(self, points: Dict[str, Tuple[int, int]]):
        self.points = points

    def dist(self, p1: str, p2: str) -> Fraction:
        # Ignores second coordinate (projection)
        v1, v2 = self.points[p1], self.points[p2]
        return Fraction(abs(v1[0] - v2[0]), 1)


space = PseudometricSpace({"A": (1, 10), "B": (1, 20)})
d_ab = space.dist("A", "B")
assert d_ab == 0
# Points are distinct in reality despite d(A, B) = 0!
points_are_equal = "A" == "B"
assert str(points_are_equal) == prediction
print("[Tested OK] Pseudometric separation falsified premature equality: dist=0 but A != B.")
````

## Part 5: The Boxed Descent Law
`[Pinned]` Objective: Verify that accepted lifts satisfy the retraction condition r_N(u_N(x)) = x.


````python
# --- PREDICTION GATE 5 ---
# For a valid lift-retraction pair (u_0, r_0), what is r_0(u_0(x))?
# Options: 'x', 'u_0(x)', 'None'
prediction = "x"


def lift_u0(x: str) -> str:
    return f"lift({x})"


def retract_r0(y: str) -> str:
    if y.startswith("lift(") and y.endswith(")"):
        return y[5:-1]
    raise ValueError("Not in lift image")


val = "<0>"
retracted = retract_r0(lift_u0(val))
assert (retracted == val) and (prediction == "x")
print(f"[Pinned OK] Boxed descent law verified: retract_r0(lift_u0({val})) == {retracted}")
````
