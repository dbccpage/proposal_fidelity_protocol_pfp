## 1. The Big Idea

**Core Concept:** LANDED, RESIDUAL, or REFUSED_EMPTY terminal outcomes

### The Four-Tier Perspective
- **Tangible (Age 5–10):** Physical intuition of terminal state trichotomy.
- **Algorithmic (Age 11–17):** Deterministic state transitions and rule sets.
- **Discrete Math (Undergrad):** Exact rational invariants and coboundary calculus.
- **Foundational (Age 50–500):** Monotone manifold retraction and topological obstruction theory.


````python
def terminal(admissible, demand):
    nxt = admissible & demand
    if not admissible:
        return "REFUSED_EMPTY"
    if len(nxt) == 1:
        return "LANDED"
    return "RESIDUAL"


result = terminal({1, 2}, {2})
expected = "LANDED"
assert terminal(set(), {1}) == "REFUSED_EMPTY"
assert terminal({1, 2}, {3}) == "RESIDUAL"
assert result == expected
print("Terminal", result)
````
