## 1. The Big Idea

**Core Concept:** Matrix representation: delta^0 v = v_j - v_i

### The Four-Tier Perspective
- **Tangible (Age 5–10):** Physical intuition of calculating coboundary matrices.
- **Algorithmic (Age 11–17):** Deterministic state transitions and rule sets.
- **Discrete Math (Undergrad):** Exact rational invariants and coboundary calculus.
- **Foundational (Age 50–500):** Monotone manifold retraction and topological obstruction theory.


````python
from fractions import Fraction

potential = (Fraction(0), Fraction(2), Fraction(5))
edges = ((0, 1), (1, 2), (2, 0))
coboundary = tuple(potential[target] - potential[source] for source, target in edges)
result = coboundary
expected = (Fraction(2), Fraction(3), Fraction(-5))
assert result == expected
assert sum(result) == 0
print("Coboundary", result)
````
