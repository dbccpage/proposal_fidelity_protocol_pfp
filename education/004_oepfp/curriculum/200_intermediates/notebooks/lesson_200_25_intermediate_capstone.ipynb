````python
from fractions import Fraction

before = (Fraction(1), Fraction(1), Fraction(1))
alpha = (Fraction(0), Fraction(1), Fraction(1))
after = (alpha[1] - alpha[0], alpha[2] - alpha[1], alpha[0] - alpha[2])
result = sum(after)
expected = Fraction(0)
assert sum(before) == 3
assert after == (Fraction(1), Fraction(0), Fraction(-1))
assert result == expected
print("Capstone circulation", sum(before), "->", result)
````
