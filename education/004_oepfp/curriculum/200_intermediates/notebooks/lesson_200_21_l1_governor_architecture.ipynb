## 1. The Big Idea

**Core Concept:** Layer-3 governor intercepting and scoring model candidates

### The Four-Tier Perspective

````python
from fractions import Fraction


def admit(current, proposed):
    return proposed <= current


result = admit(Fraction(3), Fraction(1))
expected = True
assert result is expected
assert admit(Fraction(1), Fraction(3)) is False
print("Non-increasing candidate admitted")
````
