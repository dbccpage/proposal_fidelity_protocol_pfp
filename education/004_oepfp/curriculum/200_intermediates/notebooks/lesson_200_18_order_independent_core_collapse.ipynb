## 1. The Big Idea

**Core Concept:** Proving collapse confluence regardless of operator application order

### The Four-Tier Perspective

````python
left = {1, 2, 3}
middle = {2, 3, 4}
right = {3, 9}
forward = (left & middle) & right
backward = right & (middle & left)
result = forward
expected = {3}
assert result == backward == expected
print("Order-independent core", result)
````
