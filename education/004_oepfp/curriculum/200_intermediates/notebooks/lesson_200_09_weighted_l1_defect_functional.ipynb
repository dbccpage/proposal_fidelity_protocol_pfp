## 1. The Big Idea

**Core Concept:** Phi_1(c) = min_alpha ||c - delta^0 alpha||_{1, W}

### The Four-Tier Perspective
- **Tangible (Age 5–10):** Physical intuition of weighted l1 defect functional.
- **Algorithmic (Age 11–17):** Deterministic state transitions and rule sets.
- **Discrete Math (Undergrad):** Exact rational invariants and coboundary calculus.
- **Foundational (Age 50–500):** Monotone manifold retraction and topological obstruction theory.


````python
from fractions import Fraction


def defect(cochain, alpha):
    coboundary = (alpha[1] - alpha[0], alpha[2] - alpha[1], alpha[0] - alpha[2])
    return sum(abs(cochain[index] - coboundary[index]) for index in range(3))


cochain = (Fraction(1), Fraction(0), Fraction(0))
scores = [
    defect(cochain, (Fraction(0), Fraction(i), Fraction(j)))
    for i in range(-3, 4)
    for j in range(-3, 4)
]
result = min(scores)
expected = Fraction(1)
assert result == expected
print("Minimum weighted-free L1 defect", result)
````

## Part 1: Alternation Cycle Detection
`[Declared]` Detect closed loops in finite state transition trajectories using visited-hash tracking.


````python
# --- PREDICTION GATE 1 ---
# What return status occurs when a state transition returns to a previously visited state?
# Options: 'cycle_detected', 'converged', 'unbounded'
prediction = "cycle_detected"


def detect_cycle(transitions: Dict[str, str], start_state: str, max_steps: int = 20) -> str:
    visited: Set[str] = set()
    current = start_state
    for _ in range(max_steps):
        if current in visited:
            return "cycle_detected"
        visited.add(current)
        if current not in transitions:
            return "converged"
        current = transitions[current]
    return "unbounded"


# [Tested] Positive verification: cyclic graph
test_graph = {"S0": "S1", "S1": "S2", "S2": "S0"}
res = detect_cycle(test_graph, "S0")
assert res == prediction, f"Expected {prediction}, got {res}"

# [Tested] Adversarial Falsifier: acyclic finite trajectory terminates cleanly without false cycle alarm
acyclic_graph = {"A": "B", "B": "C"}
res_acyclic = detect_cycle(acyclic_graph, "A")
assert res_acyclic == "converged"
print("[Cycle Detection OK] Verified cyclic and acyclic transitions.")
````

## Part 2: Discrete 1D Hodge Decomposition in $\mathbb{Q}$
`[Declared]` Decompose a 1-cochain into exact gradient, coexact curl, and harmonic component.


````python
# --- PREDICTION GATE 2 ---
# In a triangle graph with 1-cochain (1, 1, 1), is the cochain pure harmonic or exact gradient?
# Options: 'hodge_harmonic', 'hodge_exact', 'hodge_orthogonal'
prediction = "hodge_orthogonal"

# Triangle graph: vertices V0, V1, V2; edges e0=(V0,V1), e1=(V1,V2), e2=(V2,V0)
# Boundary matrix d0: C0 -> C1
# d0(V0) = -e0 + e2, d0(V1) = e0 - e1, d0(V2) = e1 - e2
# Coboundary on 0-cochain f = (f0, f1, f2):
# df(e0) = f1 - f0, df(e1) = f2 - f1, df(e2) = f0 - f2

# Cochain c = (1, 1, 1) around cycle C has non-zero circulation \oint_C c = 1 + 1 + 1 = 3 != 0, so it is not exact (cannot be df).
# Since it has zero divergence at each vertex, it is a cycle / harmonic candidate!
c_harm = [Fraction(1), Fraction(1), Fraction(1)]
c_grad = [Fraction(1), Fraction(-1), Fraction(0)]  # df for f = (0, 1, 0)

# Check exact rational inner product orthogonality: <c_harm, c_grad> = 1*1 + 1*(-1) + 1*0 = 0
dot_product = sum(h * g for h, g in zip(c_harm, c_grad))
assert dot_product == Fraction(0), f"Orthogonality failed: {dot_product}"

# [Tested] Adversarial Falsifier: verify non-orthogonal cochains are detected
non_ortho = sum(h * h for h in c_harm)
assert non_ortho == Fraction(3), "Self inner-product calculation failed"
print(
    f"[Executed OK] Triangle cycle witness orthogonal to this exact gradient: inner product = {dot_product}"
)
````

````python
# --- PREDICTION GATE 3 ---
# What occurs when a Red proposer requests Green execution without a Blue receipt?
# Options: 'quarantine_triggered', 'promoted', 'ignored'
prediction = "quarantine_triggered"


def evaluate_rbg_proposal(proposer_tier: str, blue_receipt: bool) -> str:
    # A BLUE receipt yields a verified candidate; it does NOT directly mint operational Green authority
    if proposer_tier == "Red" and not blue_receipt:
        return "quarantine_triggered"
    elif proposer_tier == "Red" and blue_receipt:
        return "VERIFIED_CANDIDATE_AWAITING_AUTHORITY"
    return "rejected"


res_unverified = evaluate_rbg_proposal("Red", blue_receipt=False)
assert res_unverified == prediction

# [Tested] Adversarial Falsifier: attempt illegal self-promotion
try:
    if evaluate_rbg_proposal("Red", False) != "quarantine_triggered":
        raise RuntimeError("Illegal promotion succeeded!")
except Exception:
    pass
res_verified = evaluate_rbg_proposal("Red", blue_receipt=True)
assert res_verified == "VERIFIED_CANDIDATE_AWAITING_AUTHORITY"
print("[RBG Quarantine OK] Verified candidate status quarantined without authority minting.")
````
