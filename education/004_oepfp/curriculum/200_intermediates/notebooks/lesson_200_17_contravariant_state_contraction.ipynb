## 1. The Big Idea

**Core Concept:** Monotonic shrinking of the admissible region A_{k+1} subset A_k

````python
first = {1, 2, 3, 4}
second = first & {1, 2, 3}
third = second & {1, 2}
result = (len(first), len(second), len(third))
expected = (4, 3, 2)
assert result == expected
print("Contraction counts", result)
````

## 3. Diagnostic Probe (2026 Micro-Attention Check)

**Question:** How do the admissible counts move under two nested demands?

- **A:** 4, then 3, then 2.
- **B:** They stay at 4.


````python
first = {1, 2, 3, 4}
second = first & {1, 2, 3}
third = second & {1, 2}
result = (len(first), len(second), len(third))
expected = (4, 3, 2)
assert result == expected
print("Contraction counts", result)


observed = result
assert observed == expected
print("Diagnostic Probe: PASSED", observed)
````
