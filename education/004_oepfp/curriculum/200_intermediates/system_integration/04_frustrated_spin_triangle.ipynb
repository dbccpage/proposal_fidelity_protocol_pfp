## Worked computation

Run the cell, then compare the output with the exact model above.


```python
from itertools import product

spin = load_source(
    "spin_lesson",
    "omega_engine_education/curriculum/the_symbol_game/demos/demo_4/dossier_1_f2_spin_triangle/checker.py",
)
D = [[1, 1, 0], [0, 1, 1], [1, 0, 1]]
assert spin.matmul_mod2([[1, 1, 1]], D) == [[0, 0, 0]]
weights = []
for candidate in product((0, 1), repeat=3):
    computed = spin.matvec_mod2(D, list(candidate))
    direct = [candidate[0] ^ candidate[1], candidate[1] ^ candidate[2], candidate[2] ^ candidate[0]]
    assert computed == direct
    residual = [1 ^ value for value in computed]
    assert sum(residual) % 2 == 1
    weights.append(sum(residual))
assert min(weights) == 1 and weights.count(1) == 6 and weights.count(3) == 2
print("All eight spin assignments checked; minimum residual weight is one.")
```

## Exercise and adversarial check

Change the target bonds to (1,1,0). Find an assignment with zero residual and explain the changed parity.

Predict the result before running the reference check below. Each assertion checks a specific value or failure mode.


```python
target = [1, 1, 0]
solutions = [s for s in product((0, 1), repeat=3) if spin.matvec_mod2(D, list(s)) == target]
assert set(solutions) == {(0, 1, 0), (1, 0, 1)}
broken_D = [[1, 0, 0], [0, 1, 1], [1, 0, 1]]
assert spin.matmul_mod2([[1, 1, 1]], broken_D) != [[0, 0, 0]]
print("An even target parity has two gauge-related zero-residual assignments.")
```
