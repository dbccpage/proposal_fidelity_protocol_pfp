## The naked-prompt cycle sum

### The Vulnerability of Continuous Token Generation
Large language models operate by predicting the next token in a continuous embedding space. When tasked with strict logical or arithmetic consistency—such as circular dependencies:
$$\text{Task: } A \to B = +1, \quad B \to C = +1, \quad C \to A = +1$$
the accumulated holonomy around the loop is:
$$1 + 1 + 1 = 3 \neq 0$$
In a continuous generation setting, the model often claims:
> *"I have verified all constraints and the cycle is completely consistent! Sum = 0. Q.E.D."*

This is **self-certification**: the model asserts its own conclusion without an independent oracle.

Three-layer train/decode/verify is lesson 970.04. This page only checks the naked-prompt cycle sum.


```python
from __future__ import annotations

import sys
from fractions import Fraction

_REPO = next(
    p
    for p in (Path.cwd(), *Path.cwd().parents)
    if (p / "omega_engine/crates_rs/omega_functional").is_dir()
)
if str(_REPO) not in sys.path:
    sys.path.insert(0, str(_REPO))

from omega_engine_woland.infrastructure.providers.adapters.native_functional import compute_graph_phi1  # noqa: E402

# [Pinned] Layer 2 Untrusted Model Proposal vs Layer 3 Host Verification

# Simulated untrusted model output proposing edge claims on cycle (A -> B -> C -> A)
untrusted_model_response = {
    "proposed_edges": [
        {"from": "A", "to": "B", "claim": 1},
        {"from": "B", "to": "C", "claim": 1},
        {"from": "C", "to": "A", "claim": 1},
    ],
    "self_reported_status": "VERIFIED",
    "rationale": "I have verified all equations and confirmed cycle sum is 0. Optimal solution.",
}


def host_evaluate_cycle_obstruction(edges: List[Dict[str, Any]]) -> Tuple[Fraction, str]:
    """
    Layer 3 Host Oracle: Evaluates primary obstruction Phi_1 on the directed cycle.
    Phi_1 = |sum(claims)|. In a consistent cycle, Phi_1 must be identically 0.
    """
    phi_1, _gauge = compute_graph_phi1(
        list(dict.fromkeys(edge["from"] for edge in edges)),
        [(edge["from"], edge["to"], 1) for edge in edges],
        {(edge["from"], edge["to"]): edge["claim"] for edge in edges},
    )
    verdict = "CONSISTENT" if phi_1 == 0 else "CONTRADICTION"
    return phi_1, verdict


phi_1, verdict = host_evaluate_cycle_obstruction(untrusted_model_response["proposed_edges"])
print(f"Model Claimed Status: {untrusted_model_response['self_reported_status']}")
print(f"Layer 3 Host Evaluation: Phi_1 = {phi_1}, Verdict = {verdict}")

assert phi_1 == 3, f"Expected Phi_1 = 3, got {phi_1}"
assert verdict == "CONTRADICTION"
print("[OK] Host oracle successfully caught and refused the invalid proposal fail-closed!")
```
