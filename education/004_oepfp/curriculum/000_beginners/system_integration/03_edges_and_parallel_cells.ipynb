## Start here

Draw two routes from the same start to the same finish. Their labels may differ, but a cell between them needs matching boundaries. Bracket syntax alone cannot supply this condition.

## Exact model

Parallel arrows a and b satisfy source(a)=source(b) and target(a)=target(b).

All numerical fixtures use integers or Fraction. A passing run means the named finite checks passed.


```python
import importlib.util
import sys
from pathlib import Path

ROOT = next(
    path
    for path in [Path.cwd(), *Path.cwd().parents]
    if (path / "omega_engine_education/scripts/build_system_integration.py").is_file()
)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
from omega_engine_education.scripts.system_integration_support import (
    expect_error,
)


def load_source(name, relative):
    spec = importlib.util.spec_from_file_location(name, ROOT / relative)
    module = importlib.util.module_from_spec(spec)
    sys.modules[name] = module
    spec.loader.exec_module(module)
    return module
```

## Worked computation

Run the cell, then compare the output with the exact model above.


```python
from omega_engine_pfp.domain.tower import cells, grammar


def arrow(label, target="1"):
    return cells.cell_from_packet(grammar.parse(f"op<<1011><1><{label}><0><{target}>>"))


a, b = arrow("10"), arrow("11")
assert a.id != b.id
assert cells.verify_parallel(a, b)
two = cells.cell_from_packet(
    grammar.parse("op<<1011><10><11>op<<1011><1><10><0><1>>op<<1011><1><11><0><1>>>")
)
assert cells.verify_two_cell_is_well_formed(two)
print("Distinct parallel arrows support the declared two-cell.")
```

## Exercise and adversarial check

Change the second arrow's target to 0. Predict parsing and endpoint-check outcomes separately.

Predict the result before running the reference check below. Each assertion checks a specific value or failure mode.


```python
assert not cells.verify_parallel(a, arrow("11", "0"))
bad = cells.cell_from_packet(
    grammar.parse("op<<1011><10><11>op<<1011><1><10><0><1>>op<<1011><1><11><0><0>>>")
)
assert not cells.verify_two_cell_is_well_formed(bad)
expect_error(ValueError, lambda: cells.cell_from_packet(grammar.parse("op<<1010><0><0><0><0>>")))
print("A parsed packet with mismatched endpoints fails the cell check.")
```
