## Worked computation

Run the cell, then compare the output with the exact model above.


```python
from pfp.domain.tower import grammar
from pfp.domain.tower import updown_maps as maps

base = grammar.parse("<>")
lifted = maps.u_0(base)
assert maps.d_1(lifted) == base
new_atom = grammar.parse("<1>")
receipt = maps.d_1(new_atom)
assert isinstance(receipt, maps.NonDescentReceipt)
assert receipt.original == new_atom
assert receipt != base
print("Round-trip on the old image; explicit loss record outside it.")
```
