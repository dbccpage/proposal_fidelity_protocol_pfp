## Worked computation

Run the cell, then compare the output with the exact model above.


```python
from pfp.domain.tower import grammar

examples = ["<>", "<0>", "op<<0><><1>>"]
for source in examples:
    tree = grammar.parse(source)
    assert grammar.serialize(tree) == source
assert isinstance(grammar.parse("<>"), grammar.Diff)
assert grammar.parse("<0>") != grammar.parse("<>")
print("Three distinct packet shapes round-trip.")
```

## Exercise and adversarial check

Predict which malformed strings fail. Add a nested application and remove one closing bracket.

Predict the result before running the reference check below. Each assertion checks a specific value or failure mode.


```python
for malformed in ["", "<", "<0><1>", "op<<0>"]:
    expect_error(grammar.GrammarError, lambda value=malformed: grammar.parse(value))
nested = "op<op<<0><1>><>>"
assert grammar.serialize(grammar.parse(nested)) == nested
print("Missing boundaries and trailing packets are refused.")
```
