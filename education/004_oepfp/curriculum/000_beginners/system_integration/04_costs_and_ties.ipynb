## Worked computation

Run the cell, then compare the output with the exact model above.


```python
from pfp.domain.tower import tower_policies as policies

candidates = policies.CANDIDATES
first = policies.rank_by_fewest_constructors_then_rules(candidates)
second = policies.rank_by_description_length(candidates)
tied = policies.all_tied_under(candidates, key=lambda item: item.n_constructors)
assert first == policies.PARITY_TAG
assert second == policies.LENGTH_SCHEMA
assert first != second and set(tied) == set(candidates)
print("Constructor policy and declared description cost choose different lifts.")
```

## Exercise and adversarial check

Reverse the input order. Check that the tied set is unchanged; then change one declared cost.

Predict the result before running the reference check below. Each assertion checks a specific value or failure mode.


```python
assert set(
    policies.all_tied_under(tuple(reversed(candidates)), key=lambda x: x.n_constructors)
) == set(tied)
cheaper = policies.LiftCandidate("declared_alternative", 1, 3, 7)
assert policies.rank_by_description_length(candidates + (cheaper,)) == cheaper
assert policies.rank_by_fewest_constructors_then_rules(candidates + (cheaper,)) == first
print("Ranking depends on an explicit policy and explicit data.")
```
