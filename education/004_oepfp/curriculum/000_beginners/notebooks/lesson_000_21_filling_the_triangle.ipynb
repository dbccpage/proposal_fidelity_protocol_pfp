## 1. The 2-Simplex Sheet and Simplicial Boundary Operator $\partial_2$

[Pinned] In 1D graphs, loops of edges are hollow. Now we attach a **2-simplex** $F = [0, 1, 2]$: a solid triangular sheet spanning the three boundary edges.

The oriented simplicial boundary operator on a 2-simplex $[v_0, v_1, v_2]$ is:
$$\partial_2 [v_0, v_1, v_2] = [v_1, v_2] - [v_0, v_2] + [v_0, v_1].$$
In terms of oriented cycle edges $e_0 = (0, 1)$, $e_1 = (1, 2)$, and $e_2 = (2, 0) = -(0, 2)$:
$$\partial_2 F = (0, 1) + (1, 2) + (2, 0).$$

Applying $\partial_1$ yields fundamental **boundary nilpotence**:
$$\partial_1 \circ \partial_2 = 0.$$

```python
# 1. Define oriented 2-simplex face F = [0, 1, 2]
face = (0, 1, 2)
# Boundary 1-chain: edges (0, 1), (1, 2), (2, 0) with coefficient +1
boundary_2 = {(0, 1): 1, (1, 2): 1, (2, 0): 1}

# 2. Apply 1D boundary operator d_1: (u, v) -> v - u
boundary_1 = {0: 0, 1: 0, 2: 0}
for (u, v), coeff in boundary_2.items():
    boundary_1[u] -= coeff
    boundary_1[v] += coeff

# 3. Assert exact nilpotence d_1 o d_2 = 0
assert boundary_1 == {0: 0, 1: 0, 2: 0}
assert sum(boundary_1.values()) == 0

print("Simplicial boundary d_2 and nilpotence d_1 o d_2 = 0 verified!")
```

## 2. 1-Skeleton vs. Filled 2-Complex: Non-Cocycle Status of Circulating Cochains

[Pinned] A vital topological distinction:
1. **On the 1-Skeleton Graph $G$:**
   There are no 2-cells ($C^2(G) = 0$). Hence $d^1 c = 0$ vacuously for all $c \in C^1(G)$. Every 1-cochain is vacuously a cocycle ($\ker d^1 = C^1$).
2. **On the Filled 2-Complex $K = \Delta^2$:**
   $C^2(K) = \text{span}(F)$. The coboundary evaluates by duality:
   $$d^1 c(F) = \langle c, \partial_2 F \rangle = \sum_{e \in \partial_2 F} c(e).$$
   If circulation $\langle c, z \rangle = 1 \ne 0$, then $d^1 c(F) = 1 \ne 0$.

Because $d^1 c(F) \ne 0$, **$c$ is NOT a cocycle on the filled 2-complex** ($c \notin \ker d^1$)!
The triangle $\Delta^2$ is contractible ($H^1(\Delta^2) = 0$). The non-zero value is discrete **flux / curvature** penetrating the face, NOT a surviving $H^1$ cohomology class.


```python
from fractions import Fraction as Q

# 1-cochain with circulation 1
c = {(0, 1): Q(1), (1, 2): Q(0), (2, 0): Q(0)}

# On 1-skeleton: C^2 is {0}, so d^1 c is vacuously 0
is_cocycle_1skeleton = True
assert is_cocycle_1skeleton

# On filled 2-complex: evaluate coboundary d^1 c(F)
flux_F = sum(c[e] * coeff for e, coeff in boundary_2.items())
assert flux_F == Q(1)

# Because flux_F != 0, c is NOT a cocycle in C^1(\Delta^2)!
is_cocycle_2complex = flux_F == 0
assert not is_cocycle_2complex

# Test coboundary nilpotence: exact cochain c = d^0 phi
phi = {0: Q(5), 1: Q(2), 2: Q(-1)}
c_exact = {
    (0, 1): phi[1] - phi[0],
    (1, 2): phi[2] - phi[1],
    (2, 0): phi[0] - phi[2],
}
flux_exact = sum(c_exact[e] * coeff for e, coeff in boundary_2.items())
# d^1(d^0 phi)(F) == phi(d_1(d_2 F)) == phi(0) == 0
assert flux_exact == Q(0)

print("Non-cocycle status and coboundary nilpotence d^1 o d^0 = 0 verified!")
```
