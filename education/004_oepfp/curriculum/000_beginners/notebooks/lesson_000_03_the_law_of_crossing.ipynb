## 1. The Big Idea: Involution and Original-State Identity

- **Involution:** An operator $\mathcal{I}$ satisfying $\mathcal{I}^2 = \mathrm{id}$. Double negation restores the *original* state (whether marked or unmarked).
- **Syntax vs. Semantics:** In Grammar 000 syntax, `op<op<<>>>` preserves the nested AST; it does not automatically cancel during parsing.
- **Signed Pairing:** In algebraic cancellations, pairing $+w$ and $-w$ on a closed domain with equal magnitudes drives exact cancellation ($w + (-w) = 0$).

```python
# [Pinned] Environment Preflight & Project Root Resolution
# Pinned revision: 3f3399766108adc4c4e7615dce2573a57be57dd6
# Supported environment: Python 3.12+ (e.g. py -3.12 or pixi run)
# Project install: pip install -e .
import sys
from pathlib import Path

ROOT_DIR = Path.cwd().resolve()
found = False
while ROOT_DIR.parent != ROOT_DIR:
    if (
        (ROOT_DIR / "omega_engine_pfp").exists()
        or (ROOT_DIR / ".git").exists()
        or (ROOT_DIR / "pixi.toml").exists()
    ):
        if (ROOT_DIR / "omega_engine_pfp").exists():
            found = True
            break
    ROOT_DIR = ROOT_DIR.parent

if found and str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))
elif not found:
    raise RuntimeError(
        "Could not locate 'omega_engine_pfp' root. "
        "Ensure you execute from a repository checkout or install via: pip install -e ."
    )

from omega_engine_pfp.domain.tower import grammar


# 1. Boolean involution restores original state
def boolean_involution(state: bool) -> bool:
    return not (not state)


assert boolean_involution(True) is True
assert boolean_involution(False) is False
print("Boolean involution verified: restores original state identically")

# 2. Syntax preserves nested AST (no auto-erasure)
nested = grammar.parse("op<op<<>>>")
assert isinstance(nested, grammar.App)
assert isinstance(nested.children[0], grammar.App)
assert isinstance(nested.children[0].children[0], grammar.Diff)
print("Syntactic preservation verified: nested App is preserved in AST")

# 3. Signed pairing cancellation on closed domain
weights = {"A": 5, "B": -5}
assert sum(weights.values()) == 0
print("Signed weight pairing verified: sum =", sum(weights.values()))
```
