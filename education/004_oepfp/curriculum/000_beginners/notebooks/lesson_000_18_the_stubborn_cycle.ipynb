## 1. Homology ($H_1$) vs. Cohomology ($H^1$)

[Declared] In algebraic topology, lower indices denote **chains and homology**, while upper indices denote **cochains and cohomology**:
- **Chains and Homology ($H_1$):** $C_1(G)$, boundary operator $\partial_1$, $H_1(G) = \ker \partial_1 / \text{im } \partial_2$.
- **Cochains and Cohomology ($H^1$):** $C^1(G)$, coboundary operator $d^0$, $H^1(G) = \ker d^1 / \text{im } d^0$.

## 2. Graph Non-Exactness: Why Nonzero Circulation Obstructs Potentials

[Pinned] On a 1-dimensional graph $G$ with vertices and edges (and no 2-cells):
- Every 1-cochain $c$ is vacuously a cocycle: $\ker d^1 = C^1(G)$.
- A 1-cochain $c$ is called **exact** if $c = d^0 \phi$ for some $\phi \in C^0(G; \mathbb{Q})$.
- **The Non-Exactness Theorem:** If a closed cycle $z$ has nonzero circulation $\langle c, z \rangle \ne 0$, then $c$ **cannot be exact**.

### Proof by Contradiction
Suppose there existed a potential $\phi$ such that $c = d^0 \phi$.
Then by the Telescoping Theorem: $\langle c, z \rangle = \langle d^0 \phi, z \rangle = 0$.
But for $c = (2, -1, 0)$, $\langle c, z \rangle = 1 \ne 0$. Contradiction ($0 = 1$)!

## 3. The Filled-Face Qualification

[Declared] If a 2-cell $F$ with boundary $\partial_2 F = z$ is attached to the graph:
$$d^1 c(F) = \langle c, \partial_2 F \rangle = \langle c, z \rangle = 1 \ne 0.$$
Because $d^1 c(F) \ne 0$, $c$ is **not a cocycle** on the filled 2-complex ($c \notin \ker d^1$).

```python
from fractions import Fraction as Q

# 1. Algebraic proof of non-exactness
# System:
# phi_1 - phi_0 = 2
# phi_2 - phi_1 = -1
# phi_0 - phi_2 = 0
# Sum of left sides: (phi_1 - phi_0) + (phi_2 - phi_1) + (phi_0 - phi_2) = 0
# Sum of right sides: 2 + (-1) + 0 = 1
sum_lhs = 0
sum_rhs = Q(2) + Q(-1) + Q(0)
assert sum_lhs != sum_rhs
assert sum_rhs == Q(1)

# 2. Filled-face coboundary check
# On a filled 2-cell F with boundary cycle z = (1, 1, 1):
# d^1 c(F) = <c, partial_2 F> = <c, z> = 1
d1_c_face = sum_rhs
assert d1_c_face == Q(1)
# Nonzero d^1 c means c is NOT a cocycle on the filled 2-complex!
is_cocycle_on_filled_triangle = d1_c_face == Q(0)
assert is_cocycle_on_filled_triangle is False

print("Non-exactness on 1D graph and filled-face non-cocycle status verified successfully!")
```
