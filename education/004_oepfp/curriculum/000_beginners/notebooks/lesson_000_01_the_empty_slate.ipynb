## Part 1: Parsing the Bare Boundary `<>`
`[Declared]` Objective: Verify the distinction between the unmarked medium and the formal token `<>`.


````python
# --- PREDICTION GATE 1 ---
# What AST type does parsing the token '<>' return?
# Options: 'Diff', 'Atom', 'EmptyString', 'None'
prediction = "Diff"  # Commit your hypothesis before executing

import re


@dataclass(frozen=True)
class Diff:
    """The bare distinction. Nullary token with no payload or children."""

    def __repr__(self) -> str:
        return "<>"


@dataclass(frozen=True)
class Atom:
    """Named atomic word strictly in lowercase ASCII [0-9a-z]+."""

    name: str

    def __post_init__(self):
        if not self.name or not re.fullmatch(r"[0-9a-z]+", self.name):
            raise ValueError(f"Atom payload must be non-empty ASCII [0-9a-z]+, got {self.name!r}")

    def __repr__(self) -> str:
        return f"<{self.name}>"


def parse_simple(token: str) -> Union[Diff, Atom]:
    """Parses a single self-delimiting packet under strict byte profile."""
    if not isinstance(token, str) or not token:
        raise ValueError("Packet cannot be empty")
    if token != token.strip() or any(c in token for c in "\r\n\t "):
        raise ValueError(f"Whitespace violation in strict wire profile: {token!r}")
    if token == "<>":
        return Diff()
    if token.startswith("<") and token.endswith(">"):
        payload = token[1:-1]
        if "<" not in payload and ">" not in payload:
            return Atom(payload)
    raise ValueError(f"Malformed token or unadmitted top-level sequence: {token!r}")


def parse_children_simple(inner: str) -> Tuple[Union[Diff, Atom], ...]:
    """Parses a sequence of sibling packets inside an application box op<...>."""
    if not isinstance(inner, str):
        raise ValueError("Interior must be a string")
    if any(c in inner for c in "\r\n\t "):
        raise ValueError(f"Whitespace violation in application interior: {inner!r}")
    children = []
    idx = 0
    while idx < len(inner):
        if inner[idx] != "<":
            raise ValueError(f"Expected '<' at index {idx} in {inner!r}")
        close_idx = inner.find(">", idx)
        if close_idx == -1:
            raise ValueError(f"Unclosed packet in interior: {inner!r}")
        child_str = inner[idx : close_idx + 1]
        children.append(parse_simple(child_str))
        idx = close_idx + 1
    return tuple(children)


# [Executed] Run parser and verify prediction
result = parse_simple("<>")
assert type(result).__name__ == prediction, (
    f"Prediction mismatch: expected {prediction}, got {type(result).__name__}"
)
assert isinstance(result, Diff)
print(f"[Executed OK] Returned: {result} (type={type(result).__name__})")
````

````python
# --- PREDICTION GATE 2 ---
# What AST class does '<0>' instantiate?
# Options: 'Diff', 'Atom', 'Int', 'Sequence'
prediction = "Atom"  # Commit your hypothesis

result_atom = parse_simple("<0>")
assert type(result_atom).__name__ == prediction, (
    f"Mismatch: expected {prediction}, got {type(result_atom).__name__}"
)
assert isinstance(result_atom, Atom)
assert result_atom.name == "0"
# [Tested] Structural inequality between bare distinction and atomic payload
assert type(parse_simple("<>")) is not type(result_atom)
print(
    f"[Executed OK] Bare distinction <> is distinct from payload <0>: {parse_simple('<>') != result_atom}"
)
````

## Part 5: Repository Parser Bridge & Parity Falsifiers
