## 1. The Big Idea: Self-Delimiting Siblings and Parsing Law 1

- **Parsing Law 5:** Siblings require no commas because each child packet is self-delimiting.
- **Parsing Law 1:** Input must be a single outer packet. Top-level multiple packets `<0><1>` are rejected; siblings must be grouped inside `op<<0><1>>`.
- **Round-Trip:** Establishing `serialize(parse(x)) == x` for arbitrary nesting requires structural induction over grammar productions.

```python
# [Pinned] Environment Preflight & Project Root Resolution
# Pinned revision: 3f3399766108adc4c4e7615dce2573a57be57dd6
# Supported environment: Python 3.12+ (e.g. py -3.12 or pixi run)
# Project install: pip install -e .
import sys
from pathlib import Path

ROOT_DIR = Path.cwd().resolve()
found = False
while ROOT_DIR.parent != ROOT_DIR:
    if (
        (ROOT_DIR / "pfp").exists()
        or (ROOT_DIR / ".git").exists()
        or (ROOT_DIR / "pixi.toml").exists()
    ):
        if (ROOT_DIR / "pfp").exists():
            found = True
            break
    ROOT_DIR = ROOT_DIR.parent

if found and str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))
elif not found:
    raise RuntimeError(
        "Could not locate 'pfp' root. "
        "Ensure you execute from a repository checkout or install via: pip install -e ."
    )

from pfp.domain.tower import grammar

# 1. Sibling packets inside an App (Parsing Law 5)
app = grammar.parse("op<<0><1>>")
assert isinstance(app, grammar.App)
assert len(app.children) == 2
assert [c.word for c in app.children] == ["0", "1"]
print("Parsed siblings inside App:", app)

# 2. Parsing Law 1: Multiple top-level outer packets are rejected
try:
    grammar.parse("<0><1>")
    assert False, "Multiple outer packets must raise GrammarError"
except grammar.GrammarError as e:
    assert "more than one outer packet" in e.reason.lower()
    print("Top-level siblings without App correctly rejected:", e.reason)
```
