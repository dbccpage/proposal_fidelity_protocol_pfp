## 1. The Big Idea: Polygons, Boundaries, and 1-Cycles

- **Nondegenerate planar polygon:** In the Euclidean plane $\mathbb{R}^2$, a simple straight-edged polygon requires at least 3 vertices and 3 straight line segments to enclose a finite area.
- **Boundary 1-Skeleton vs. Filled 2-Simplex:**
  - The boundary 1-skeleton (perimeter) of a triangle has Euler characteristic $\chi = V - E = 3 - 3 = 0$.
  - The filled triangle (including interior 2-cell) has $\chi = V - E + F = 3 - 3 + 1 = 1$.
  - A compact enclosed region like an annulus also has $\chi = 0$; compact enclosure does not require nonzero Euler characteristic.
- **Cycle Orientation:** To form a 1-cycle $z = \sum e_i$ with $\partial_1 z = 0$, edges must be consistently oriented ($0 \to 1 \to 2 \to 0$).

```python
# Lesson 000.02: Drawing a Boundary
# 1. Euler characteristics of boundary 1-skeleton vs filled triangle
chi_skeleton = 3 - 3
chi_filled = 3 - 3 + 1
assert chi_skeleton == 0
assert chi_filled == 1
print("Boundary skeleton chi =", chi_skeleton, "(loop / annulus)")
print("Filled triangle chi =", chi_filled, "(simply connected 2-cell)")

# 2. Simplicial boundary operator on oriented triangle cycle z = e0 + e1 + e2
edges = [(0, 1), (1, 2), (2, 0)]
boundary = {0: 0, 1: 0, 2: 0}
for u, v in edges:
    boundary[u] -= 1
    boundary[v] += 1
assert boundary == {0: 0, 1: 0, 2: 0}
print("Oriented boundary operator vanishes exactly:", boundary)
```

## 4. Student Exercise

**Task:** Verify that an unoriented or reversed edge breaks boundary vanishing $\partial_1 z = 0$.

```python
# Exercise: Verify that reversing one edge breaks boundary vanishing
misoriented_edges = [(0, 1), (1, 2), (0, 2)]
mis_boundary = {0: 0, 1: 0, 2: 0}
for u, v in misoriented_edges:
    mis_boundary[u] -= 1
    mis_boundary[v] += 1
assert mis_boundary != {0: 0, 1: 0, 2: 0}
assert mis_boundary == {0: -2, 1: 0, 2: 2}
print("Exercise verified: unoriented loop fails boundary vanishing!")
```
