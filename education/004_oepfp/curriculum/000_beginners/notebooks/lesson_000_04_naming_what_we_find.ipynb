## 1. The Big Idea: Letters, Words, and Nonempty Atoms

- **Alphabet & Words:** The alphanumeric alphabet $\Sigma_{36} = \{0..9, a..z\}$ generates words by concatenation.
- **Nonempty Atom Payloads:** An `Atom` must have a nonempty payload word. Empty brackets `<>` are reserved for `Diff()`.
- **Recursive Syntax:** Packets form the recursive sum type `Diff | Atom | App`.

```python
# [Pinned] Environment Preflight & Project Root Resolution
# Pinned revision: 3f3399766108adc4c4e7615dce2573a57be57dd6
# Supported environment: Python 3.12+ (e.g. py -3.12 or pixi run)
# Project install: pip install -e .
import sys
from pathlib import Path

ROOT_DIR = Path.cwd().resolve()
found = False
while ROOT_DIR.parent != ROOT_DIR:
    if (
        (ROOT_DIR / "pfp").exists()
        or (ROOT_DIR / ".git").exists()
        or (ROOT_DIR / "pixi.toml").exists()
    ):
        if (ROOT_DIR / "pfp").exists():
            found = True
            break
    ROOT_DIR = ROOT_DIR.parent

if found and str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))
elif not found:
    raise RuntimeError(
        "Could not locate 'pfp' root. "
        "Ensure you execute from a repository checkout or install via: pip install -e ."
    )

from pfp.domain.tower import grammar

# 1. Parse valid atoms
atom0 = grammar.parse("<0>")
atoma = grammar.parse("<a>")
assert isinstance(atom0, grammar.Atom) and atom0.word == "0"
assert isinstance(atoma, grammar.Atom) and atoma.word == "a"
print("Parsed valid atoms:", atom0, atoma)

# 2. Empty atom is rejected: must use Diff() for <>
try:
    grammar.Atom("")
    assert False, "Expected ValueError on empty Atom"
except ValueError as e:
    assert "nonempty" in str(e).lower()
    print("Empty atom rejected:", e)
```
