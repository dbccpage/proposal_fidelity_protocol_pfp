## 3. Interactive Graph Frustration Explorer

Use the sliders below to adjust edge claims around the triangle cycle $A \to B \to C$.
Watch how defects are highlighted in red, and click the projection button to find the minimum $L_1$ defect.


```python
import sys
from fractions import Fraction
from pathlib import Path

sys.path.insert(0, str(Path(".").resolve()))

from notebook_lib import GovernorAdapter, InteractiveWidgets

vertices = ["A", "B", "C"]
edges = [("A", "B", 1), ("B", "C", 1), ("A", "C", 1)]

# Consistent cycle: A->B (1) + B->C (2) == A->C (3)
claims_consistent = {
    ("A", "B"): Fraction(1),
    ("B", "C"): Fraction(2),
    ("A", "C"): Fraction(3),
}
res_cons = GovernorAdapter.project_l1_exact(vertices, edges, claims_consistent)
print(
    f"[Pinned] Consistent Graph: Phi_1 = {res_cons.primary_obstruction}, Consistent = {res_cons.is_cycle_consistent}"
)

# Frustrated cycle: A->B (1) + B->C (2) != A->C (0)
claims_frustrated = {
    ("A", "B"): Fraction(1),
    ("B", "C"): Fraction(2),
    ("A", "C"): Fraction(0),
}
res_frust = GovernorAdapter.project_l1_exact(vertices, edges, claims_frustrated)
print(
    f"[Pinned] Frustrated Graph: Phi_1 = {res_frust.primary_obstruction}, Consistent = {res_frust.is_cycle_consistent}"
)
```
