## Section 1: Simplicial Complexes & Graded Cochain Algebra ($C^0, C^1, C^2$)

### [Declared] Graded Cochain Complex Definition
Let $K = (V, E, F)$ be a finite 2-truncated cell complex over exact rational coefficients $\mathbb{Q}$.
- $C^0(K; \mathbb{Q}) = \mathbb{Q}^V$ is the space of 0-cochains (vertex potentials $f: V \to \mathbb{Q}$).
- $C^1(K; \mathbb{Q}) = \mathbb{Q}^E$ is the space of 1-cochains (edge claims $h: E \to \mathbb{Q}$), with anti-symmetry $h(v, u) = -h(u, v)$.
- $C^2(K; \mathbb{Q}) = \mathbb{Q}^F$ is the space of 2-cochains (face curvatures $\omega: F \to \mathbb{Q}$).

### [Declared] Graded Coboundary Operators
1. The **0-coboundary** $d^0: C^0 \to C^1$ assigns to each directed edge $e = (u, v)$ the potential difference:
   $$(d^0 f)(u, v) = f(v) - f(u)$$
2. The **1-coboundary** $d^1: C^1 \to C^2$ evaluates the oriented circulation around each triangular 2-cell $F = (u, v, w)$:
   $$(d^1 h)(u, v, w) = h(v, w) - h(u, w) + h(u, v)$$

### [Pinned] Nilpotency on one complex

The general identity is background. This cell checks one finite complex. It does not decide the repair-versus-quotient question.
For any vertex potential $f \in C^0$, substitution gives:
$$(d^1 (d^0 f))(u, v, w) = (f(w) - f(v)) - (f(w) - f(u)) + (f(v) - f(u)) = 0$$
This nilpotency holds algebraically over $\mathbb{Q}$ with exact zero defect.


```python
# [Pinned] Nilpotency on one complex

# 1. Complete tournament complex K_3 with 1 face
K3 = SimplicialComplex.complete_graph(3, with_faces=True)
assert K3.num_vertices == 3
assert K3.num_edges == 3
assert K3.num_faces == 1

# Test arbitrary rational vertex potential f
f_arbitrary = Cochain0({"v0": Fraction(13, 7), "v1": Fraction(-4, 3), "v2": Fraction(42, 5)})
h_edge = d0(K3, f_arbitrary)
face_defect = d1(K3, h_edge)

# The face curvature must vanish identically
assert face_defect.get("v0", "v1", "v2") == Fraction(0)
print(f"[OK] Face curvature d^1(d^0 f): {face_defect.get('v0', 'v1', 'v2')}")

# 2. Audit nilpotency across all elementary basis vectors
receipt_basis = audit_nilpotency_all_basis(K3)
assert receipt_basis.is_nilpotent is True
assert receipt_basis.max_absolute_defect == Fraction(0)
print(f"[OK] Complete basis nilpotency audit: {receipt_basis.details}")

# 3. Higher complex: Complete graph K_4 with 4 faces
K4 = SimplicialComplex.complete_graph(4, with_faces=True)
receipt_k4 = audit_nilpotency_all_basis(K4)
assert receipt_k4.is_nilpotent is True
print(f"[OK] K_4 basis nilpotency audit over {K4.num_faces} faces: {receipt_k4.details}")
```
