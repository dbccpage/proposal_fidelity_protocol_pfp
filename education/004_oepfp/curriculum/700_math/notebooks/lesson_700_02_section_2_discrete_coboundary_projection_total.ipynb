## One weighted deviation on a triangle

The weighted absolute deviation at one potential is the sum inside Phi_1. Phi_1 itself is the minimum of that sum. This cell evaluates the sum at one chosen potential on one triangle. It does not compute the minimum.

Claims on `AB`, `BC`, and `CA` are each 1. The potential is `A = 0`, `B = 1`, `C = 2`. The sum at that potential is 3.

Lesson 700.06 reads the native quotient. Lesson 700.10 checks subdeterminants of one incidence matrix. `domain.ddc.compute_phi1` refuses until a Rust adapter is wired, so this cell does not call it and does not build a certificate.

The Poincaré / Hoffman-Kruskal statement for every directed graph is background. This cell does not decide it.


```python
# [Pinned] Import verified DDC domain package and numeric toolchains
from fractions import Fraction

```

```python
claims = {("A", "B"): Fraction(1), ("B", "C"): Fraction(1), ("C", "A"): Fraction(1)}
weights = {edge: Fraction(1) for edge in claims}
potential = {"A": Fraction(0), "B": Fraction(1), "C": Fraction(2)}

deviation = Fraction(0)
for (tail, head), claim in claims.items():
    deviation += weights[(tail, head)] * abs(claim - (potential[head] - potential[tail]))

assert deviation == Fraction(3)
print(deviation)
```
