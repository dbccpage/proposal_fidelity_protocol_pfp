## 1. The Anatomy of Conversational Fluff ("Bullshit") in LLMs

Unconstrained language models generate substantial extraneous tokens:
- **Preamble & Pleasantries**: *"Certainly! I would be delighted to assist you..."*
- **Sycophancy & False Confidence**: *"Clearly", "obviously", "as we all know", "trivially"*.
- **Smuggled Ontology**: Invoking continuous manifolds, Hilbert spaces, or quantum terms in discrete algebraic tasks.
- **Unpriced Rewrites**: Intermediate assertions made without proof citations.

### The Adversary Swarm Operational Axiom
> *"An unpriced rewrite is a smuggled axiom."*

The Symbol Game enforces a **Support Restriction Operator** $\mathcal{P}_K: \Delta(V) \to \Delta(K)$, acting as an $L_1$ convex relaxation of $L_0$ token support sparsity.


```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path(".").resolve()))

from notebook_lib import SymbolGameLLMClient

client = SymbolGameLLMClient(mode="mock")

dense_text = (
    "Certainly! I would be delighted to help. First, let us explore the rich algebraic properties. "
    "Clearly and obviously, by simple inspection of the continuous continuum limit, the manifold behaves nicely. "
    "In conclusion, verified: true. Q.E.D. Hope this helps!"
)

sparse_text = (
    "ASSERT edge_1 -> residue_0\nCUT edge_2\nVERIFY primary_obstruction == 0\nLAND verified"
)

res_dense = client.analyze_fluff(dense_text, is_symbol_game=False)
res_sparse = client.analyze_fluff(sparse_text, is_symbol_game=True)

print("=== Fluff Analysis Comparison ===")
print(
    f"Dense CoT   -> Tokens: {res_dense.token_count}, Fluff: {res_dense.fluff_count}, Smuggled: {res_dense.smuggled_count}, Unpriced Density: {res_dense.unpriced_density}"
)
print(
    f"Symbol Game -> Tokens: {res_sparse.token_count}, Fluff: {res_sparse.fluff_count}, Smuggled: {res_sparse.smuggled_count}, Unpriced Density: {res_sparse.unpriced_density}"
)
```
