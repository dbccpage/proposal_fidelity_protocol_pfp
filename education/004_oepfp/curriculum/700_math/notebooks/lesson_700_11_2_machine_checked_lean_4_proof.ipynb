Every column of $\partial_1$ has exactly one $+1$ and one $-1$, ensuring $\sum_v (\partial_1)_{v, e} = 0$.


```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path(".").resolve()))

from notebook_lib import GovernorAdapter

# Incidence matrix of an oriented 3-cycle
incidence_matrix = [
    [-1, 0, 1],  # Node A
    [1, -1, 0],  # Node B
    [0, 1, -1],  # Node C
]

is_tum = GovernorAdapter.check_total_unimodularity(incidence_matrix)
print(f"[Pinned] Incidence matrix is Totally Unimodular (TUM): {is_tum}")
```

```python
# Verify column sum invariant computationally
n_rows = len(incidence_matrix)
n_cols = len(incidence_matrix[0])

for c in range(n_cols):
    col_sum = sum(incidence_matrix[r][c] for r in range(n_rows))
    print(
        f"[Pinned] Column {c} sum = {col_sum} (Matches Lean 4 TotalUnimodularity.column_sum_zero)"
    )
    assert col_sum == 0
```

## 4. Student Exercises

### Exercise 06.1: Submatrix Determinants
Why does Total Unimodularity ensure that solving a system of linear equations $A x = b$ with integer coefficients $A$ and $b$ produces integer solutions?

<details>
<summary><b>Click to reveal solution</b></summary>
By Cramer's rule, the components of the solution $x$ are ratios of determinants: $x_i = \det(A_i) / \det(A)$.
If $A$ is totally unimodular, any non-singular basis submatrix has $\det(A) = \pm 1$.
Since $\det(A_i)$ is an integer and $\det(A) = \pm 1$, the quotient is always an exact integer.
</details>
