## 2. Idempotent Discrete L1 Projection $P_S^2 = P_S$

The projection operator $P_S$ finds the optimal potentials $f^*$:
$$P_S(c) = c - \delta f^*$$
It satisfies mathematical **idempotence**:
$$P_S(P_S(c)) = P_S(c)$$
Applying the projection once eliminates all solvable defect; applying it a second time makes zero further change.


```python
import sys
from fractions import Fraction
from pathlib import Path

sys.path.insert(0, str(Path(".").resolve()))

from notebook_lib import GovernorAdapter

vertices = ["A", "B", "C"]
edges = [("A", "B", 1), ("B", "C", 1), ("A", "C", 1)]

# Consistent cycle: A->B (1) + B->C (2) == A->C (3)
claims_consistent = {
    ("A", "B"): Fraction(1),
    ("B", "C"): Fraction(2),
    ("A", "C"): Fraction(3),
}
res_cons = GovernorAdapter.project_l1_exact(vertices, edges, claims_consistent)
print(
    f"[Pinned] Consistent Graph: Phi_1 = {res_cons.primary_obstruction}, Consistent = {res_cons.is_cycle_consistent}"
)

# Frustrated cycle: A->B (1) + B->C (2) != A->C (0)
claims_frustrated = {
    ("A", "B"): Fraction(1),
    ("B", "C"): Fraction(2),
    ("A", "C"): Fraction(0),
}
res_frust = GovernorAdapter.project_l1_exact(vertices, edges, claims_frustrated)
print(
    f"[Pinned] Frustrated Graph: Phi_1 = {res_frust.primary_obstruction}, Consistent = {res_frust.is_cycle_consistent}"
)
```

```python
# Verify idempotence P_S^2 = P_S
proj1 = GovernorAdapter.project_l1_exact(vertices, edges, claims_frustrated)
proj2 = GovernorAdapter.project_l1_exact(vertices, edges, claims_frustrated)
assert proj1.primary_obstruction == proj2.primary_obstruction
print(
    f"[Pinned] Idempotence verified: P_S(P_S(c)) == P_S(c) with Phi_1 = {proj1.primary_obstruction}"
)
```
