This sitting checks subdeterminants of one directed complete bipartite graph $K_{2,3}$.


```python
# [Pinned] Import verified DDC domain package and numeric toolchains
import itertools

import numpy as np

```

```python
# [Pinned] Subdeterminants of one bipartite incidence matrix

# Solution 1: Bipartite Graph K_{2,3} Incidence Matrix TUM Check
U_nodes = ["u0", "u1"]
V_nodes = ["v0", "v1", "v2"]
bipartite_edges = [(u, v) for u in U_nodes for v in V_nodes]  # 6 edges
all_nodes = U_nodes + V_nodes  # 5 vertices

M_bip = np.zeros((len(bipartite_edges), len(all_nodes)), dtype=int)
for i, (u, v) in enumerate(bipartite_edges):
    M_bip[i, all_nodes.index(v)] = 1
    M_bip[i, all_nodes.index(u)] = -1

# Check subdeterminants
bip_tum = True
for k in range(1, 6):
    for r_comb in itertools.combinations(range(len(bipartite_edges)), k):
        for c_comb in itertools.combinations(range(len(all_nodes)), k):
            sub = M_bip[np.ix_(r_comb, c_comb)]
            det = int(round(np.linalg.det(sub)))
            if det not in (-1, 0, 1):
                bip_tum = False
                break
assert bip_tum is True
print("[OK] Exercise 1 verified: Directed bipartite graph K_{2,3} is totally unimodular!")
```
