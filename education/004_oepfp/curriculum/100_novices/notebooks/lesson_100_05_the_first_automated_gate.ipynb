````python
# Lesson 100.05: The First Automated Gate
def gate(pkt: dict) -> dict:
    if not pkt.get("carrier_set") or not pkt.get("payload"):
        return {"status": "REFUSED", "code": "REFUSED_EMPTY", "admitted": False}
    return {"status": "ADMITTED", "code": "OK", "admitted": True}


res = gate({"carrier_set": "MSS-1"})
print("Gate evaluation result:", res)
assert res["status"] == "REFUSED"
print("Deterministic rejection oracle verified!")
````

## Part 1: Blue Syntax Firewall & Delimiter Verification
`[Declared]` Validate self-closing wire delimiters before invoking semantic evaluators.


````python
# --- PREDICTION GATE 1 ---
# What return status occurs when a wire string with unclosed brackets '<0><1' hits the syntax firewall?
# Options: 'REFUSED_SYNTAX_DEFECT', 'ADMITTED', 'QUARANTINED'
prediction = "REFUSED_SYNTAX_DEFECT"


def syntax_firewall_filter(wire: str) -> str:
    if not wire:
        return "REFUSED_EMPTY_PAYLOAD"
    # Bracket balancing and self-closing delimiter check
    stack = 0
    for char in wire:
        if char == "<":
            stack += 1
        elif char == ">":
            if stack <= 0:
                return "REFUSED_SYNTAX_DEFECT"
            stack -= 1
        elif stack == 0 and char.strip():
            return "REFUSED_UNWRAPPED_GLYPH"
    if stack != 0:
        return "REFUSED_SYNTAX_DEFECT"
    return "SYNTAX_PASS"


status_bad = syntax_firewall_filter("<0><1")
assert status_bad == prediction

# [Tested] Positive fixture: well-formed wire string passes syntax firewall
assert syntax_firewall_filter("<0><1>") == "SYNTAX_PASS"
print("[Syntax Firewall OK] Verified delimiter enforcement and unclosed bracket rejection.")
````

## Part 2: Terminal Status Reachability Table
`[Declared]` Classify evaluated proposals into the canonical 5 terminal statuses.


````python
# --- PREDICTION GATE 2 ---
# What terminal status is assigned to a syntactically valid proposal that requires a higher lift license?
# Options: 'LiftLicenseNeeded', 'Admitted', 'Refused'
prediction = "LiftLicenseNeeded"

TERMINAL_STATUSES = {"Admitted", "Refused", "Quarantined", "Deferred", "LiftLicenseNeeded"}


def evaluate_pfp_candidate(current_stage: int, target_stage: int, payload_valid: bool) -> str:
    if not payload_valid:
        return "Refused"
    if target_stage > current_stage + 1:
        return "LiftLicenseNeeded"
    if target_stage == current_stage + 1:
        return "Admitted"
    return "Deferred"


res_license = evaluate_pfp_candidate(current_stage=2, target_stage=5, payload_valid=True)
assert res_license == prediction
assert res_license in TERMINAL_STATUSES

# [Tested] Adversarial Falsifier: invalid payload is strictly Refused
assert evaluate_pfp_candidate(2, 3, payload_valid=False) == "Refused"
print(
    f"[Terminal Table OK] Verified reachability status: {res_license} in declared terminal statuses."
)
````

## Part 3: Red Proposal Quarantine & Zero Authority Effect
`[Declared]` Enforce that candidate proposals cannot self-promote to operational authority without external verification.


````python
# --- PREDICTION GATE 3 ---
# Does a Red proposer's verified receipt directly grant operational authority?
# Options: 'VERIFIED_CANDIDATE_AWAITING_AUTHORITY', 'DIRECT_EXECUTION', 'AUTONOMOUS_PROMOTION'
prediction = "VERIFIED_CANDIDATE_AWAITING_AUTHORITY"


class PFPProposal:
    def __init__(self, proposer_id: str, candidate_data: str):
        self.proposer_id = proposer_id
        self.candidate_data = candidate_data
        self.authority_effect = "None"


def process_proposal_pipeline(prop: PFPProposal, checker_passed: bool) -> str:
    # A checker pass validates correctness; it does NOT grant operational authority
    if not checker_passed:
        return "QUARANTINE_REJECTED"
    if prop.authority_effect != "None":
        return "ERR_ILLEGAL_AUTHORITY_INJECTION"
    return "VERIFIED_CANDIDATE_AWAITING_AUTHORITY"


prop = PFPProposal(proposer_id="agent_red_01", candidate_data="<state_delta_42>")
status = process_proposal_pipeline(prop, checker_passed=True)
assert status == prediction

# [Tested] Adversarial Falsifier: proposal attempting to assert non-zero authority is rejected
prop_bad = PFPProposal(proposer_id="agent_red_02", candidate_data="<exploit>")
prop_bad.authority_effect = "Operational"
assert process_proposal_pipeline(prop_bad, checker_passed=True) == "ERR_ILLEGAL_AUTHORITY_INJECTION"
print("[Authority Quarantine OK] Verified candidate status quarantined with AuthorityEffect: None.")
````
