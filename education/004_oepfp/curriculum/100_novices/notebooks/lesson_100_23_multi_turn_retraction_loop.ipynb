## Integrated System Integration: A potential repair leaves cycle sum unchanged (Coboundary Cycle Repair)


AuthorityEffect: None.

## Start here

Assign a height to each vertex. Each edge records the difference between its endpoint heights. Walking around a closed triangle cancels these differences, so such a repair cannot change the cycle sum.

## Exact model

h' = h - d0(f); sum_cycle d0(f) = 0, hence sum_cycle h' = sum_cycle h.

All numerical fixtures use integers or Fraction. A passing run means the named finite checks passed.


````python
import importlib.util
import sys
from fractions import Fraction
from pathlib import Path

ROOT = next(
    path
    for path in [Path.cwd(), *Path.cwd().parents]
    if (path / "omega_engine_education/scripts/build_system_integration.py").is_file()
)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
from omega_engine_education.scripts.system_integration_support import (
    expect_error,
)


def load_source(name, relative):
    spec = importlib.util.spec_from_file_location(name, ROOT / relative)
    module = importlib.util.module_from_spec(spec)
    sys.modules[name] = module
    spec.loader.exec_module(module)
    return module
````

## Worked computation

Run the cell, then compare the output with the exact model above.


````python
from omega_engine_pfp.domain.ddc import Cochain0, Cochain1, Edge, SimplicialComplex, d0

edges = (Edge("a", "b"), Edge("b", "c"), Edge("c", "a"))
carrier = SimplicialComplex(("a", "b", "c"), edges)
h = Cochain1({edge.key: Fraction(1, 3) for edge in edges})
potential = Cochain0({"a": Fraction(0), "b": Fraction(1, 3), "c": Fraction(2, 3)})
repair = d0(carrier, potential)
residual = h - repair
assert sum(repair.values.values(), Fraction(0)) == 0
assert sum(residual.values.values(), Fraction(0)) == 1
assert list(residual.values.values()).count(Fraction(1)) == 1
assert sum(abs(v) for v in residual.values.values()) == 1
print("The repair concentrates the unit cycle defect; it does not erase it.")
````

## Exercise and adversarial check

Shift every vertex height by 7/5. Then negate the original edge values and compare the cycle sum.

Predict the result before running the reference check below. Each assertion checks a specific value or failure mode.


````python
shifted = Cochain0({key: value + Fraction(7, 5) for key, value in potential.values.items()})
assert d0(carrier, shifted) == repair
assert sum((-h - repair).values.values(), Fraction(0)) == -1
expect_error(ValueError, lambda: Edge("a", "a"))
print("Constant shifts preserve the repair; reversing the defect changes its signed cycle sum.")
````
