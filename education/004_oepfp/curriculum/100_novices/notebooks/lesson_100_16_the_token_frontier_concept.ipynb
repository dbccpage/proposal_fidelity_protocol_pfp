````python
from fractions import Fraction

# Lesson 100.16: The Token Frontier Concept
grammar_state = "IN_TAG"
vocab_dist = {"<": 1, ">": 6, "a": 2, " ": 1}
frontier = {">", "a"}
retained = sum(Fraction(vocab_dist[t], 10) for t in frontier)
print("Active State:", grammar_state, "| Frontier:", frontier)
print("Retained probability mass M =", retained)
assert retained == Fraction(8, 10)
````

## Start here

A grammar masks disallowed tokens. The remaining probabilities must be divided by their retained mass; if no mass remains, decoding needs a refusal instead of an unmasked fallback.

## Exact model

For S={0,2} and weights (1,2,3,4), M=2/5 and conditioned weights are (1/4,3/4).

All numerical fixtures use integers or Fraction. A passing run means the named finite checks passed.


````python
import importlib.util
import sys
from fractions import Fraction
from pathlib import Path

ROOT = next(
    path
    for path in [Path.cwd(), *Path.cwd().parents]
    if (path / "education/scripts/build_system_integration.py").is_file()
)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))


def load_source(name, relative):
    spec = importlib.util.spec_from_file_location(name, ROOT / relative)
    module = importlib.util.module_from_spec(spec)
    sys.modules[name] = module
    spec.loader.exec_module(module)
    return module
````

## Worked computation

Run the cell, then compare the output with the exact model above.


````python
frontier = load_source(
    "frontier_lesson",
    "pfp/infrastructure/adapters/mssn_scenarios/token_frontier.py",
)
model = frontier.MockModelLogits(("a", "b", "c", "d"), (1, 2, 3, 4))
result = frontier.evaluate_token_frontier_step(model, [0, 2])
assert result.is_ok()
retained = result.unwrap()
def to_fraction(r):
    return Fraction(r.numerator, r.denominator)
assert to_fraction(retained.retained_mass.mass) == Fraction(2, 5)
probabilities = [to_fraction(item.constrained_probability) for item in retained.candidates]
assert probabilities == [Fraction(1, 4), Fraction(3, 4)]
assert sum(probabilities, Fraction(0)) == 1
print(
    "Retained mass:",
    to_fraction(retained.retained_mass.mass),
    "conditioned:",
    list(map(str, probabilities)),
)
````

## Exercise and adversarial check

Try empty support, a zero-weight token, and a single positive-weight token. Explain the outcomes.
