````python
import hashlib
import json

# Lesson 100.13: SHA-256 Envelope Sealing
data = {"carrier": "MSS-1", "move": "CUT(v=1)", "turn": 3}
raw_bytes = json.dumps(data, sort_keys=True, separators=(",", ":")).encode("utf-8")
seal = hashlib.sha256(raw_bytes).hexdigest()
print("Generated SHA-256 seal:", seal[:20] + "...")
assert len(seal) == 64
print("Envelope sealed and verifiable against tamper!")
````
