## 3. Diagnostic Probe (2026 Micro-Attention Check)

**Question:** Will swapping the two example keys change their encoded bytes?

Choose `same_bytes` or `different_bytes` before running the comparison.
The filled prediction below is an author example, not learner evidence.


````python
prediction = "same_bytes"  # Change before running the next cell.
````

````python
first = {"b": 2, "a": 1}
second = {"a": 1, "b": 2}
expected = b'{"a":1,"b":2}'
assert prediction == "same_bytes"
assert canonical_json_bytes(first) == expected
assert canonical_json_bytes(second) == expected
assert canonical_json_bytes({"a": 3, "b": 2}) != expected
assert canonical_json_bytes({"text": "two words"}) == b'{"text":"two words"}'
print("Worked finite byte comparisons passed")
````

````python
assert canonical_json_bytes({"a": "é"}) == '{"a":"é"}'.encode("utf-8")
# Default number behavior is a limit, not strict-ingress enforcement.
assert canonical_json_bytes({"n": float("nan")}) == b'{"n":NaN}'
rejected = False
try:
    canonical_json_bytes({"value": object()})
except TypeError:
    rejected = True
assert rejected
print("Finite boundary checks passed; no full-format or learner claim")
````
