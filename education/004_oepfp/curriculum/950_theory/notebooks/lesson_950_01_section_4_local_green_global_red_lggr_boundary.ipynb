## Section 4: Local Green / Global Red (LGGR) Boundary Dynamics

### [Declared] The Asymmetric Verdict Law
The central epistemological principle of the Symbol Game and CODA is the **Asymmetric Verdict Law**:

```python
from adversary.infrastructure.providers.adapters.native_functional import compute_graph_phi1

# [Proved] LGGR Boundary Dynamics Simulation: Local Green, Global Red


class LGGREvaluator:
    """
    Evaluates local subgraphs versus global union complex.
    Demonstrates that local green verdicts do not authorize global green certification.
    """

    @classmethod
    def evaluate_complex(
        cls,
        global_complex: SimplicialComplex,
        local_subgraphs: List[SimplicialComplex],
        claim: Cochain1,
    ) -> Dict[str, Any]:
        # 1. Evaluate local patches
        local_results = []
        for idx, sub in enumerate(local_subgraphs):
            # Extract sub-claim
            sub_vals = {e.key: claim.get(e.u, e.v) for e in sub.edges}
            Cochain1(sub_vals)
            phi_1, _gauge = compute_graph_phi1(
                list(sub.vertices),
                [(edge.u, edge.v, edge.weight) for edge in sub.edges],
                {(edge.u, edge.v): claim.get(edge.u, edge.v) for edge in sub.edges},
            )
            local_status = "GREEN" if phi_1 == Fraction(0) else "RED"
            local_results.append(
                {"subgraph_id": f"Sub_{idx}", "phi_1": phi_1, "status": local_status}
            )

        # 2. Evaluate global complex
        global_claim = {
            (edge.u, edge.v): claim.get(edge.u, edge.v) for edge in global_complex.edges
        }
        global_phi_1, _gauge = compute_graph_phi1(
            list(global_complex.vertices),
            [(edge.u, edge.v, edge.weight) for edge in global_complex.edges],
            global_claim,
        )
        global_status = "GREEN" if global_phi_1 == Fraction(0) else "RED"

        # 3. Apply Asymmetric Verdict Law
        all_local_green = all(r["status"] == "GREEN" for r in local_results)
        asymmetric_conflict = all_local_green and (global_status == "RED")

        verdict_code = "LOCAL_GREEN_GLOBAL_RED" if asymmetric_conflict else global_status

        return {
            "all_local_green": all_local_green,
            "global_status": global_status,
            "global_phi_1": global_phi_1,
            "asymmetric_conflict": asymmetric_conflict,
            "final_verdict": verdict_code,
            "local_reports": local_results,
        }


# Construct 3 local path graphs (each is a tree with 0 cycles -> Phi_1 is always 0)
# Subgraph 1: v0 -> v1
# Subgraph 2: v1 -> v2
# Subgraph 3: v2 -> v0
Sub1 = SimplicialComplex(vertices=("v0", "v1"), edges=(Edge("v0", "v1"),), faces=())
Sub2 = SimplicialComplex(vertices=("v1", "v2"), edges=(Edge("v1", "v2"),), faces=())
Sub3 = SimplicialComplex(vertices=("v2", "v0"), edges=(Edge("v2", "v0"),), faces=())

# Global complex: union of all three forms a frustrated 3-cycle
K_global = SimplicialComplex.cycle_graph(3)

# Edge claims: (+1, +1, +1)
claim_lggr = Cochain1(
    {("v0", "v1"): Fraction(1), ("v1", "v2"): Fraction(1), ("v2", "v0"): Fraction(1)}
)

lggr_result = LGGREvaluator.evaluate_complex(K_global, [Sub1, Sub2, Sub3], claim_lggr)

print("Local Subgraph Evaluations:")
for rep in lggr_result["local_reports"]:
    print(f"  {rep['subgraph_id']}: Phi_1 = {rep['phi_1']}, Status = {rep['status']}")

print(
    f"\nGlobal Complex Evaluation: Phi_1 = {lggr_result['global_phi_1']}, Status = {lggr_result['global_status']}"
)
print(f"Asymmetric Verdict Law Outcome: {lggr_result['final_verdict']}")

assert lggr_result["all_local_green"] is True, (
    "Every tree subgraph must evaluate to Phi_1 = 0 (Green)"
)
assert lggr_result["global_status"] == "RED", (
    "The macroscopic global cycle must evaluate to Phi_1 = 3 (Red)"
)
assert lggr_result["asymmetric_conflict"] is True
assert lggr_result["final_verdict"] == "LOCAL_GREEN_GLOBAL_RED"
print(
    "[OK] Asymmetric Verdict Law successfully demonstrated: Local Green does NOT imply Global Green!"
)
```
