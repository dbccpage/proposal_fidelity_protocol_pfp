# Lesson 420.08: Lineage Taint Fixpoint Workbench

**AuthorityEffect:** None.  
**Level:** Level 08 (Lineage Taint & Custody Invalidation)  

This workbench computes the forward causal taint fixpoint.

```python
def compute_forward_taint(dag_children, tainted_root):
    tainted_set = set([tainted_root])
    queue = [tainted_root]
    while queue:
        curr = queue.pop(0)
        for child in dag_children.get(curr, []):
            if child not in tainted_set:
                tainted_set.add(child)
                queue.append(child)
    return tainted_set


dag = {"P0": ["P1", "P2"], "P1": ["P1_1"], "P2": ["P2_1"]}
tainted = compute_forward_taint(dag, "P1")
assert "P1_1" in tainted and "P2" not in tainted
print(f"Forward Taint Fixpoint from P1: {tainted}")
```
