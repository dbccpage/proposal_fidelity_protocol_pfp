# Lesson 420.07: Largest-Remainder Population Allocation Workbench

**Level:** Level 07 (Ecological Reserves & Population)  

This workbench verifies discrete population quota allocation across Frontier (60%), Unmeasured (25%), and Baseline (15%).

```python
import math
from fractions import Fraction


def allocate_reserves(pop_cap):
    # Quotas: Frontier 3/5, Unmeasured 1/4, Baseline 3/20
    fractions = [Fraction(3, 5), Fraction(1, 4), Fraction(3, 20)]
    raw = [float(f * pop_cap) for f in fractions]
    floored = [int(math.floor(r)) for r in raw]
    remainder = pop_cap - sum(floored)
    remainders = [(raw[i] - floored[i], i) for i in range(len(raw))]
    remainders.sort(reverse=True)
    for k in range(remainder):
        floored[remainders[k][1]] += 1
    return floored


alloc = allocate_reserves(20)
assert sum(alloc) == 20
print(f"Population Cap 20 Allocation [Frontier, Unmeasured, Baseline]: {alloc}")
```
