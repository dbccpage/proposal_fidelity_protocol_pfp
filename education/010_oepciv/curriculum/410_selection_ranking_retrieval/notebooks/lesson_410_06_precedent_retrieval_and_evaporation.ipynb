This workbench verifies epoch-based half-life decay calculation.

```python
def evaporate_weight(initial_weight, elapsed_epochs, half_life=5):
    return initial_weight * (0.5 ** (elapsed_epochs / half_life))


w0 = 1.0
w5 = evaporate_weight(w0, 5, half_life=5)
w10 = evaporate_weight(w0, 10, half_life=5)
assert w5 == 0.5
assert w10 == 0.25
print(f"Epoch 0: {w0}, Epoch 5: {w5}, Epoch 10: {w10}")
```
