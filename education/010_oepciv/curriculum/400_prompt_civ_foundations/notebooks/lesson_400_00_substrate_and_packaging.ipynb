# Lesson 400.00: Governed Prompt Material vs Execution Occurrence Workbench

**AuthorityEffect:** None.  
**Level:** Level 00 (Prompt Substrate & Packaging)  

This workbench provides interactive verification of `PromptGenomeV1` hashing and `ExecutionOccurrenceV1` independence.

```python
import hashlib
import json


def compute_prompt_id(source_packet_refs, domain_model, parents, birth_custody):
    payload = {
        "birth_custody": birth_custody,
        "domain_model": domain_model,
        "parents": parents,
        "source_packet_refs": source_packet_refs,
    }
    serialized = json.dumps(payload, sort_keys=True, separators=(",", ":"))
    return hashlib.sha256(serialized.encode("utf-8")).hexdigest()


# Preflight verification
pid1 = compute_prompt_id(["ref1"], "model_v1", [], "Wild")
pid2 = compute_prompt_id(["ref1"], "model_v1", [], "Wild")
assert pid1 == pid2, "Genome SHA256 must be deterministic"
print(f"PromptGenomeV1 SHA256: {pid1}")
```
