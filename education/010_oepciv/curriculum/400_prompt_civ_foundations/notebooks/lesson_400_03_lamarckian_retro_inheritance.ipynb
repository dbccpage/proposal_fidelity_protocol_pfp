This workbench verifies backward attenuation weight calculation: $W(d) = (1/2)^d$ for $d \le 10$.

```python
def calculate_attenuated_weight(hops, base_weight=1.0, max_hops=10):
    if hops > max_hops:
        return 0.0
    return base_weight * ((0.5) ** hops)


assert calculate_attenuated_weight(0) == 1.0
assert calculate_attenuated_weight(1) == 0.5
assert calculate_attenuated_weight(3) == 0.125
assert calculate_attenuated_weight(11) == 0.0
print("Hop-Bounded Attenuation Weights Verified.")
```
