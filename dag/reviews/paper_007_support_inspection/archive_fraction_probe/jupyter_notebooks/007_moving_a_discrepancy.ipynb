# Moving a discrepancy without erasing it

Companion notebook for *Designing executable lessons with explicit evidence boundaries*, Section 3 (version 0.1.0, 2026-10-02). It repeats the worked triangle lesson with Python's `fractions` module only and imports no repository code. Section 3 of the guide runs the same arithmetic through the repository's cochain implementation.

Draw vertices a, b, c with directed edges a to b, b to c, and c to a. Each edge carries a rational claim. A potential assigns a height to each vertex; the difference on an edge is the target height minus the source height.
Record each first prediction and reason before running its reference cell. Keep corrections separately; this notebook does not collect participant responses.


```python
from fractions import Fraction as Q

EDGES = (("a", "b"), ("b", "c"), ("c", "a"))


def d0(heights):
    """Difference on each directed edge: target height minus source height."""
    return {(s, t): heights[t] - heights[s] for (s, t) in EDGES}


def cycle_sum(values):
    """Signed sum of edge values around the cycle a -> b -> c -> a."""
    return sum((values[edge] for edge in EDGES), Q(0))


def show(values):
    return "(" + ", ".join(str(values[edge]) for edge in EDGES) + ")"


h = {edge: Q(1, 3) for edge in EDGES}
assert cycle_sum(h) == 1
print(f"claims h = {show(h)}; cycle sum = {cycle_sum(h)}")
```

## Worked computation

Choose heights f(a) = 0, f(b) = 1/3, f(c) = 2/3. Compute the edge differences df and subtract them from the claims.

```python
f = {"a": Q(0), "b": Q(1, 3), "c": Q(2, 3)}
df = d0(f)
residual = {edge: h[edge] - df[edge] for edge in EDGES}

assert [df[edge] for edge in EDGES] == [Q(1, 3), Q(1, 3), Q(-2, 3)]
assert [residual[edge] for edge in EDGES] == [Q(0), Q(0), Q(1)]
assert cycle_sum(df) == 0
assert cycle_sum(residual) == cycle_sum(h)
lines = [
    f"df = {show(df)}; cycle sum = {cycle_sum(df)}",
    f"h - df = {show(residual)}; cycle sum = {cycle_sum(residual)}",
]
print("\n".join(lines))
```

The discrepancy has moved to the edge from c to a, but its cycle sum is still one. Adding the three endpoint differences cancels every vertex height, so the sum of h - df equals the sum of h.

## Prediction before the next cell

Add 7/5 to every vertex height. Predict the new edge differences before running the cell.

```python
shifted = {vertex: height + Q(7, 5) for vertex, height in f.items()}
assert d0(shifted) == df
heights = ", ".join(str(shifted[vertex]) for vertex in "abc")
print(f"shifted heights = ({heights}); d0(shifted) = {show(d0(shifted))}")
```

Both endpoints of every edge receive the same shift, so each difference is unchanged.

Now negate the claims and keep the same potential. Predict the new signed residual sum.

```python
negated = {edge: -h[edge] - df[edge] for edge in EDGES}
assert cycle_sum(negated) == -1
print(f"-h - df = {show(negated)}; cycle sum = {cycle_sum(negated)}")
```

The answer changes because the edge claims changed, not because constant height shifts acquired a new effect.

## Completion exercise

Keep the first two claims at 1/3 and choose the third so that the cycle sum is zero.

```python
third = -(h[("a", "b")] + h[("b", "c")])
completed = {**h, ("c", "a"): third}
assert cycle_sum(completed) == 0
print(f"third claim = {third}; claims = {show(completed)}; cycle sum = {cycle_sum(completed)}")
```

## A deliberate wrong-sign control

Predict whether a difference taken as source minus target still passes the cycle-sum check. Then compute its three coordinates. Which check distinguishes it from the declared target-minus-source convention?


```python
wrong_df = {(s, t): f[s] - f[t] for s, t in EDGES}
wrong_residual = {edge: h[edge] - wrong_df[edge] for edge in EDGES}
assert cycle_sum(wrong_df) == 0
assert cycle_sum(wrong_residual) == cycle_sum(h)
assert [wrong_df[edge] for edge in EDGES] != [Q(1, 3), Q(1, 3), Q(-2, 3)]
print(f"wrong-sign df = {show(wrong_df)}; residual = {show(wrong_residual)}")
print("weak cycle-sum check passes; predicted-coordinate check rejects")
```

## Evidence boundary

These cells check one declared rational triangle and the probes above. The telescoping identity explains the cycle invariant. Neither establishes a physical interpretation, neural behavior, or a learner's mastery, and no learning, attention, or transfer outcome is measured here. A record of what a learner understood would need the learner's own predictions and an explicit assessment procedure.
The wrong-sign control shows that preserving the cycle sum alone does not establish the directed-edge convention. No learner mastery follows from these stored reference answers.
