# Demo 3: Deposited debate trace replay

This notebook replays the deposited trace offline and makes no provider call. Metrics are recomputed against historical verifier labels; they are not independent physical ground truth. Missing or duplicate conditions are refused. The separate live demo (`code/live_adversarial_demo.py`) is optional and needs the reader's own OpenRouter key.

```python
import contextlib
import io
import sys
from pathlib import Path

paper_dir = next((p for p in (Path.cwd(), Path.cwd().parent) if (p / "code" / "certificate_gate.py").exists()), None)
if paper_dir is None:
    raise RuntimeError("Start the notebook from the paper directory or its jupyter_notebooks directory")
sys.path.insert(0, str(paper_dir / "code"))


@contextlib.contextmanager
def one_output_block():
    """Send a cell's printed lines as one stream message.

    Timed flushes could otherwise split output into several messages, and a
    stored notebook would then differ from a fresh run under load.
    """
    buffer = io.StringIO()
    try:
        with contextlib.redirect_stdout(buffer):
            yield
    finally:
        sys.stdout.write(buffer.getvalue())
        sys.stdout.flush()


from run_adversarial_debate_benchmark import run_benchmark

with one_output_block():
    receipt = run_benchmark()
```

The receipt reports both historical statuses and current verification. Original model outputs remain unchanged in the deposited JSONL trace.
