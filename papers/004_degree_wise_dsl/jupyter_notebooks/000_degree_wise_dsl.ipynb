```python
from fractions import Fraction
from pathlib import Path
import sys

# Ensure local code directory is accessible
sys.path.insert(0, str(Path.cwd() / 'code'))

from mod_009_degree_wise_dsl_core import (
    create_c3_complex,
    DegreeConfusionError,
    MissingComplex2Error,
    Gamma2Status,
    DegreeWiseDSLState,
    evaluate_guarded_descent,
)

# 1. Verify coboundary nilpotency d1 ∘ d0 = 0
c3_cell = create_c3_complex(with_two_cell=True)
nilpotent = c3_cell.verify_nilpotency()
print(f"Coboundary nilpotency d1 ∘ d0 = 0: {nilpotent}")

# 2. Test action on 0-cochain f = (3, -2, 5)
f = (Fraction(3), Fraction(-2), Fraction(5))
d0_f = c3_cell.d0(f)
d1_d0_f = c3_cell.d1(d0_f)
print(f"0-cochain f: {f}")
print(f"d0(f): {d0_f}")
print(f"d1(d0(f)): {d1_d0_f}")
```

```python
# 3. Demonstrate Degree Separation on C3
omega = (Fraction(1), Fraction(0), Fraction(0))

# Graph-only model (no 2-cell attached)
g_only = create_c3_complex(with_two_cell=False)
phi1_g = g_only.compute_phi1(omega)
readout_g = g_only.compute_gamma2(omega)
print(f"Graph-only Phi1: {phi1_g}")
print(f"Graph-only Gamma2 status: {readout_g.status.name}, value: {readout_g.value}")

# Cellular complex with 2-cell attached
c_cell = create_c3_complex(with_two_cell=True)
phi1_c = c_cell.compute_phi1(omega)
readout_c = c_cell.compute_gamma2(omega)
print(f"Cellular Phi1: {phi1_c}")
print(f"Cellular Gamma2 status: {readout_c.status.name}, value: {readout_c.value}")
print(f"Degree separation demonstrated: Phi1 is identical ({phi1_g} == {phi1_c}), but Gamma2 requires 2-cell data.")
```

```python
# 4. Demonstrate Guarded Descent Evaluator Hierarchy (Phi1 > Gamma2 > Xi)
curr_metrics = (Fraction(2), Fraction(1), Fraction(5))

# Candidate improving Phi1 is accepted even if Gamma2 or Xi degrade
cand_better_phi = (Fraction(1), Fraction(2), Fraction(10))
res_phi = evaluate_guarded_descent(curr_metrics, cand_better_phi)
print(f"Candidate with improved Phi1 accepted: {res_phi}")
assert res_phi is True

# Candidate worsening Phi1 is strictly rejected even if Gamma2 and Xi improve to 0
cand_worse_phi = (Fraction(3), Fraction(0), Fraction(0))
res_worse = evaluate_guarded_descent(curr_metrics, cand_worse_phi)
print(f"Candidate with worsened Phi1 rejected: {not res_worse}")
assert res_worse is False

# 5. DSL State Representation
dsl_state = DegreeWiseDSLState(
    carrier_name="C3",
    edge_cochain=omega,
    has_two_cell=True,
    two_cell_incidence=((1, 1, 1),),
)
print(f"Constructed DSL State: carrier={dsl_state.carrier_name}, has_two_cell={dsl_state.has_two_cell}")
```
