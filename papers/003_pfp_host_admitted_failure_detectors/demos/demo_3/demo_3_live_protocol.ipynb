## Route 2: Exact Rational Cochains (Cokernel Obstruction)

Evaluates exact rational chain complexes $C^0 \xrightarrow{d_0} C^1 \xrightarrow{d_1} C^2$:
- **Exact Repair**: Finds 0-cochain potential $s$ such that $d_0 s = c_1$.
- **Cokernel Obstruction**: In fixture `cochain_nonexact_cycle`, $d_1 c_1 = 1 \neq 0$ and $H^1 = \ker(d_1)/\operatorname{im}(d_0) = 0$. The separating left-null functional $\lambda = (1, 1, 1)$ satisfies $\lambda d_0 = 0$ and $\lambda c_1 = 1 \neq 0$, certifying non-exactness ($c_1 \notin \operatorname{im}(d_0)$) in the cokernel quotient $Q_1 = C^1 / \operatorname{im}(d_0)$.

```python
# Example 2A: Exact Cochain Repair (Arm T)
res_exact = run_scripted_arm_protocol("cochain_exact_cycle", "T")
print("Exact Cochain Exit:", res_exact["final_exit"])
print("Exact Cochain Terminal:", res_exact["terminal_label"])

# Example 2B: Cokernel Non-Exact 1-Cochain Obstruction (Arm T)
res_nonexact = run_scripted_arm_protocol("cochain_nonexact_cycle", "T")
print("\nNon-exact 1-Cochain Exit:", res_nonexact["final_exit"])
print("Non-exact 1-Cochain Terminal:", res_nonexact["terminal_label"])
```

## Atomic Batch Rollback Demonstration

Demonstrates that an invalid move (e.g. attempting `VERIFY` with zero element coverage, or appending moves after `VERIFY`) triggers an immediate fail-stop rollback with zero state mutation.

```python
engine = Demo3HostEngine("gauge_even_triangle", "T")

# Attempt VERIFY with zero coverage moves -> Must be rejected with full rollback
bad_receipt = engine.execute_batch([
    MoveProposal("PROPOSE_WITNESS", {"witness_type": "vertex_bits", "witness": [0, 1, 0]}),
    MoveProposal("VERIFY", {}),
])
print("Premature Verify Decision:", bad_receipt.host_decision)
print("Batch Rolled Back:", bad_receipt.batch_rolled_back)
print("Covered Elements after Rollback:", list(engine.state.covered_elements))
print("Terminal Exit State:", engine.state.terminal_exit)
```
