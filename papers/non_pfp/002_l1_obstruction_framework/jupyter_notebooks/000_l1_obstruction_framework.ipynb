```python
from fractions import Fraction
from pathlib import Path
import sys

# Ensure local code directory is accessible
sys.path.insert(0, str(Path.cwd() / 'code'))

from mod_004_l1_obstruction_framework_core import (
    DirectedGraph,
    ExactRationalL2LaplacianSolver,
    ExactRationalLinfChebyshevSolver,
)

# Construct 3-cycle graph complex C3
graph = DirectedGraph(
    vertices=['v0', 'v1', 'v2'],
    edges=[('v0', 'v1'), ('v1', 'v2'), ('v2', 'v0')],
)

# Test exact uniform cycle cochain
omega = [Fraction(1, 1), Fraction(1, 1), Fraction(1, 1)]
l2_sol = ExactRationalL2LaplacianSolver.solve(graph, omega)

print(f'L2 Status: {l2_sol.status}')
print(f'Squared Norm: {l2_sol.squared_norm}')
print(f'Primal Potential: {l2_sol.primal_potential}')
print(f'Harmonic Residual: {l2_sol.harmonic_residual}')
print(f'Orthogonality Verified: {l2_sol.orthogonality_verified}')
```
