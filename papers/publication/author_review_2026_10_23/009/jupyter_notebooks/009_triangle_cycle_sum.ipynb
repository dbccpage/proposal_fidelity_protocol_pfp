# Triangle cycle sum

Companion to Section 3 of *Reviewing claims against code and evidence* (Paper 009).

The triangle has vertices 0, 1 and 2 and oriented edges (0, 1), (1, 2) and (2, 0).
The claims are h = (1/3, 1/3, 1/3). A potential f assigns a value to each vertex,
and its edge differences are df(u, v) = f(v) - f(u).

All values are exact fractions. The functions come from `code/triangle_cycle_sum.py`
in this package; nothing outside the package is imported.

Predict the three sums before running. Assertions check the displayed values;
output comparison checks replay consistency, not independent mathematical truth.

```python
import sys
from pathlib import Path

for base in (Path.cwd(), Path.cwd().parent):
    if (base / "code" / "triangle_cycle_sum.py").is_file():
        sys.path.insert(0, str(base / "code"))
        break
else:
    raise FileNotFoundError("Run from the package root or from jupyter_notebooks/")
```

```python
from fractions import Fraction

from triangle_cycle_sum import cycle_sum, edge_differences, subtract


def show(values):
    return "(" + ", ".join(str(value) for value in values) + ")"
```

## Reading 1: subtract the potential differences

Section 3 chooses f = (0, 1/3, 2/3). Subtracting df from the claims moves the
discrepancy onto one edge. The cycle sum does not change.

```python
claims = (Fraction(1, 3), Fraction(1, 3), Fraction(1, 3))
potential = (Fraction(0), Fraction(1, 3), Fraction(2, 3))
differences = edge_differences(potential)
residual = subtract(claims, differences)
print("h        =", show(claims))
print("df       =", show(differences))
print("h - df   =", show(residual))
print("sum(h)      =", cycle_sum(claims))
print("sum(h - df) =", cycle_sum(residual))
assert differences == (Fraction(1, 3), Fraction(1, 3), Fraction(-2, 3))
assert residual == (0, 0, 1)
assert cycle_sum(claims) == cycle_sum(residual) == 1
```

Every potential gives edge differences that sum to zero around the cycle,
because each vertex value is added once and subtracted once. Shifting every
vertex by the same amount leaves df unchanged.

```python
shifted = tuple(value + Fraction(7, 5) for value in potential)
print("sum(df)             =", cycle_sum(differences))
print("df after the shift  =", show(edge_differences(shifted)))
print("unchanged           =", edge_differences(shifted) == differences)
assert cycle_sum(differences) == 0
assert edge_differences(shifted) == differences
```

## Reading 2: replace the claims

A different operation replaces the claims by df. The new claims have cycle sum
zero because the declared input changed, not because a potential removed the
original discrepancy.

```python
replacement = differences
print("new claims      =", show(replacement))
print("sum(new claims) =", cycle_sum(replacement))
print("original sum    =", cycle_sum(claims))
assert cycle_sum(replacement) == 0
assert cycle_sum(claims) == 1
assert replacement != claims
```
